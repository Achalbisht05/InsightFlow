## Payment Analysis

* The Payments dataset contains **103,886 payment records** across **99,440 unique orders**.
* The dataset contains **5 columns** and has **no missing values**.
* `credit_card` is the dominant payment method with **76,795 payment records (73.92%)**, followed by `boleto` with **19,784 (19.04%)**.
* The average payment value is approximately **154.10 BRL**, while the median is **100.00 BRL**, indicating a right-skewed distribution caused by higher-value transactions.
* The maximum recorded payment value is **13,664 BRL**.
* Credit-card payments contribute the largest share of total payment value, with approximately **12.54 million BRL** in payment records.
* The average number of installments is approximately **2.85**, while the median is **1 installment**. The maximum is **24 installments**.
* Most orders have a single payment record (**96,479 orders**), while some orders contain multiple payment records.
* One order contains as many as **29 payment records**, demonstrating that the Payments table can have multiple rows per order.
* One order, `bfbd0f9bdef84302105ad712db648a6c`, exists in the Orders table but has **no corresponding payment record**.
* Payment analysis will be used later to investigate **payment methods, transaction value, installment behavior, and customer purchasing patterns in Power BI**.

```python
# ============================================
# RECREATE DELIVERY ANALYSIS
# ============================================

delivery_analysis = orders.copy()

# Convert date columns
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    delivery_analysis[col] = pd.to_datetime(
        delivery_analysis[col],
        errors="coerce"
    )

# Delivery time in days
delivery_analysis["delivery_days"] = (
    delivery_analysis["order_delivered_customer_date"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

# Estimated delivery time in days
delivery_analysis["estimated_delivery_days"] = (
    delivery_analysis["order_estimated_delivery_date"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

# Delivery delay
delivery_analysis["delivery_delay_days"] = (
    delivery_analysis["delivery_days"]
    - delivery_analysis["estimated_delivery_days"]
)

print("Delivery analysis shape:", delivery_analysis.shape)

display(
    delivery_analysis[
        [
            "order_id",
            "customer_id",
            "order_status",
            "delivery_days",
            "estimated_delivery_days",
            "delivery_delay_days"
        ]
    ].head()
)
```

```python
# ============================================
# 9.5 RATING VS DELIVERY PERFORMANCE
# ============================================

review_delivery = reviews[
    ["order_id", "review_score"]
].merge(
    delivery_analysis[
        ["order_id", "delivery_days", "delivery_delay_days"]
    ],
    on="order_id",
    how="left"
)

review_delivery["delivery_performance"] = np.select(
    [
        review_delivery["delivery_delay_days"] > 0,
        review_delivery["delivery_delay_days"] <= 0
    ],
    [
        "Late",
        "Early"
    ],
    default="Unknown"
)

review_delivery_analysis = (
    review_delivery
    .groupby("delivery_performance")
    .agg(
        reviews=("order_id", "nunique"),
        average_rating=("review_score", "mean"),
        average_delivery_days=("delivery_days", "mean"),
        average_delay_days=("delivery_delay_days", "mean")
    )
    .round(2)
    .sort_values("average_rating", ascending=False)
)

review_delivery_analysis
```

```python
# ============================================
# 9.6 RATING DISTRIBUTION BY DELIVERY
# ============================================

rating_delivery_matrix = pd.crosstab(
    review_delivery["delivery_performance"],
    review_delivery["review_score"]
)

rating_delivery_matrix
```

```python
# Rating percentage within each delivery group

rating_delivery_percentage = (
    pd.crosstab(
        review_delivery["delivery_performance"],
        review_delivery["review_score"],
        normalize="index"
    ) * 100
).round(2)

rating_delivery_percentage
```

```python
# ============================================
# 9.7 MONTHLY REVIEW TREND
# ============================================

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

reviews["year_month"] = (
    reviews["review_creation_date"]
    .dt.to_period("M")
    .astype(str)
)

monthly_review_analysis = (
    reviews
    .groupby("year_month")
    .agg(
        reviews=("review_id", "nunique"),
        average_rating=("review_score", "mean"),
        positive_reviews=(
            "review_score",
            lambda x: (x >= 4).sum()
        ),
        negative_reviews=(
            "review_score",
            lambda x: (x <= 2).sum()
        )
    )
)

monthly_review_analysis["positive_rate"] = (
    monthly_review_analysis["positive_reviews"]
    / monthly_review_analysis["reviews"]
    * 100
).round(2)

monthly_review_analysis["negative_rate"] = (
    monthly_review_analysis["negative_reviews"]
    / monthly_review_analysis["reviews"]
    * 100
).round(2)

monthly_review_analysis["average_rating"] = (
    monthly_review_analysis["average_rating"]
    .round(2)
)

monthly_review_analysis = monthly_review_analysis.reset_index()

monthly_review_analysis
```

```python
# ============================================
# 9.8 REVIEW COMMENT ANALYSIS
# ============================================

# Create comment length
reviews["comment_length"] = (
    reviews["review_comment_message"]
    .fillna("")
    .astype(str)
    .str.len()
)

# Create word count
reviews["comment_word_count"] = (
    reviews["review_comment_message"]
    .fillna("")
    .astype(str)
    .str.split()
    .str.len()
)

# Average comment length by rating
comment_length_by_score = (
    reviews
    .groupby("review_score")
    .agg(
        reviews=("review_id", "nunique"),
        average_comment_length=("comment_length", "mean"),
        median_comment_length=("comment_length", "median"),
        average_word_count=("comment_word_count", "mean")
    )
    .round(2)
)

comment_length_by_score
```

```python
# ============================================
# 9.8.1 REVIEW COMMENT AVAILABILITY
# ============================================

reviews["has_comment"] = (
    reviews["review_comment_message"]
    .fillna("")
    .str.strip()
    .ne("")
)

comment_availability = (
    reviews
    .groupby("review_score")
    .agg(
        total_reviews=("review_id", "nunique"),
        reviews_with_comment=("has_comment", "sum")
    )
)

comment_availability["comment_rate"] = (
    comment_availability["reviews_with_comment"]
    / comment_availability["total_reviews"]
    * 100
).round(2)

comment_availability
```

```python
# ============================================
# 9.8.2 MOST COMMON REVIEW WORDS
# ============================================

from collections import Counter
import re

review_text = " ".join(
    reviews.loc[
        reviews["has_comment"],
        "review_comment_message"
    ].astype(str)
).lower()

words = re.findall(r"\b[a-z]{3,}\b", review_text)

common_words = (
    pd.DataFrame(
        Counter(words).most_common(30),
        columns=["word", "count"]
    )
)

common_words
```

```python
# ============================================
# 9.9 FINAL REVIEW SUMMARY
# ============================================

review_summary = pd.DataFrame({
    "metric": [
        "Total Reviews",
        "Average Rating",
        "Positive Review Rate",
        "Negative Review Rate",
        "1-Star Reviews",
        "5-Star Reviews",
        "Reviews With Comments",
        "Average Comment Length - 1 Star",
        "Average Comment Length - 5 Star",
        "Early Delivery Average Rating",
        "Late Delivery Average Rating"
    ],
    "value": [
        reviews["review_id"].nunique(),
        round(reviews["review_score"].mean(), 2),
        round((reviews["review_score"] >= 4).mean() * 100, 2),
        round((reviews["review_score"] <= 2).mean() * 100, 2),
        (reviews["review_score"] == 1).sum(),
        (reviews["review_score"] == 5).sum(),
        reviews["has_comment"].sum(),
        round(
            reviews.loc[
                reviews["review_score"] == 1,
                "comment_length"
            ].mean(), 2
        ),
        round(
            reviews.loc[
                reviews["review_score"] == 5,
                "comment_length"
            ].mean(), 2
        ),
        review_delivery_analysis.loc[
            "Early", "average_rating"
        ],
        review_delivery_analysis.loc[
            "Late", "average_rating"
        ]
    ]
})

review_summary
```

## Review Analysis

* The Reviews dataset contains **99,224 review records**.
* There are **no missing values** in `review_id`, `order_id`, `review_score`, `review_creation_date`, or `review_answer_timestamp`.
* `review_comment_title` has **87,656 missing values**, while `review_comment_message` has **58,247 missing values**.
* Customer ratings range from **1 to 5**, with **5-star reviews being the most common** at 57.78%.
* The overall average review score is **4.09 out of 5**, indicating generally positive customer satisfaction.
* **77.07%** of reviews are positive (4–5 stars), **8.24%** are neutral (3 stars), and **14.69%** are negative (1–2 stars).
* Only **41.30%** of reviews contain written comments, while **58.70%** contain only a rating.
* Negative reviews are more likely to contain written feedback: **74.72%** of negative reviews had comments compared with **34.69%** of positive reviews.
* Among reviews containing comments, **1-star reviews had the longest average comments (99.94 characters)**, compared with **51.84 characters** for 5-star reviews.
* The median review response time was **40.20 hours**, while the average was **75.58 hours**, indicating a right-skewed response-time distribution caused by some very long response times.
* The **24–48 hour** response group had the highest average rating at **4.24**, while reviews answered within 24 hours had an average rating of **3.89**. This represents an association, not necessarily a causal relationship.
* Review volume increased substantially over time, with **2018 accounting for 56,164 reviews** compared with 42,735 in 2017.
* **August 2018** recorded the highest monthly review volume with **8,987 reviews**.
* There are **814 duplicate `review_id` values**, but **no completely duplicated rows**. These duplicate IDs should be investigated rather than automatically removed.
* All review scores are valid values between **1 and 5**.

### Key Business Insight

Overall customer satisfaction was strong, with an average rating of **4.09/5** and **77.07% positive reviews**. However, negative reviews are particularly valuable for improvement because customers giving low ratings were much more likely to provide detailed written feedback.

```python
import unicodedata

def normalize_city(city):
    city = str(city).lower().strip()
    city = unicodedata.normalize("NFKD", city)
    city = "".join(
        c for c in city
        if not unicodedata.combining(c)
    )
    return city

geolocation["city_normalized"] = (
    geolocation["geolocation_city"]
    .apply(normalize_city)
)
```

# Delivery / Logistics Analysis

```python
delivery_analysis = orders.copy()

# Convert date columns to datetime
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    delivery_analysis[col] = pd.to_datetime(
        delivery_analysis[col],
        errors="coerce"
    )

# Actual delivery time in days
delivery_analysis["delivery_days"] = (
    delivery_analysis["order_delivered_customer_date"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

# Approval time in days
delivery_analysis["approval_days"] = (
    delivery_analysis["order_approved_at"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

# Carrier handling time in days
delivery_analysis["carrier_days"] = (
    delivery_analysis["order_delivered_carrier_date"]
    - delivery_analysis["order_approved_at"]
).dt.total_seconds() / (24 * 60 * 60)

# Estimated delivery time in days
delivery_analysis["estimated_delivery_days"] = (
    delivery_analysis["order_estimated_delivery_date"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

# Delivery delay
delivery_analysis["delivery_delay_days"] = (
    delivery_analysis["order_delivered_customer_date"]
    - delivery_analysis["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

delivery_analysis.head()
```

```python
delivery_analysis[
    [
        "order_id",
        "order_status",
        "delivery_days",
        "approval_days",
        "carrier_days",
        "estimated_delivery_days",
        "delivery_delay_days"
    ]
].head(10)
```

```python
# ============================================
# 10.5 MISSING VALUE ANALYSIS
# ============================================

missing_report = []

for name, df in clean_datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        if missing_count > 0:

            missing_report.append({
                "dataset": name,
                "column": column,
                "missing_count": missing_count,
                "missing_percentage": round(
                    missing_count / len(df) * 100,
                    2
                )
            })

missing_report = (
    pd.DataFrame(missing_report)
    .sort_values(
        ["dataset", "missing_percentage"],
        ascending=[True, False]
    )
)

missing_report
```

```python
# ============================================
# 10.11 DUPLICATE REVIEW ID INVESTIGATION
# ============================================

# Count rows per review_id
review_id_counts = (
    reviews_clean
    .groupby("review_id")
    .size()
    .sort_values(ascending=False)
)

print("Total review rows:", len(reviews_clean))
print("Unique review IDs:", reviews_clean["review_id"].nunique())
print("Duplicated review IDs:", (review_id_counts > 1).sum())
print("Rows belonging to duplicated IDs:", review_id_counts[review_id_counts > 1].sum())

# Check whether duplicated review IDs represent
# different orders
duplicate_review_structure = (
    reviews_clean[
        reviews_clean["review_id"].isin(
            review_id_counts[review_id_counts > 1].index
        )
    ]
    .groupby("review_id")
    .agg(
        rows=("review_id", "size"),
        unique_orders=("order_id", "nunique"),
        unique_scores=("review_score", "nunique"),
        unique_comments=("review_comment_message", "nunique"),
        unique_creation_dates=("review_creation_date", "nunique")
    )
)

display(
    duplicate_review_structure
    .sort_values(
        ["unique_orders", "rows"],
        ascending=False
    )
    .head(20)
)
```

```python
# ============================================
# 10.11.1 CLASSIFY DUPLICATE REVIEW IDs
# ============================================

duplicate_structure_summary = pd.Series({
    "Same ID + Same Order":
        (
            (duplicate_review_structure["unique_orders"] == 1)
            & (duplicate_review_structure["rows"] > 1)
        ).sum(),

    "Same ID + Different Orders":
        (
            duplicate_review_structure["unique_orders"] > 1
        ).sum(),

    "Same ID + Different Scores":
        (
            duplicate_review_structure["unique_scores"] > 1
        ).sum(),

    "Same ID + Different Comments":
        (
            duplicate_review_structure["unique_comments"] > 1
        ).sum()
})

display(duplicate_structure_summary)
```

```python
# ============================================
# 10.12 DUPLICATE REVIEW IDs WITH DIFFERENT ORDERS
# ============================================

review_id_different_orders = (
    duplicate_review_structure[
        duplicate_review_structure["unique_orders"] > 1
    ]
)

print(
    "Review IDs reused across different orders:",
    len(review_id_different_orders)
)

display(
    reviews_clean[
        reviews_clean["review_id"].isin(
            review_id_different_orders.index
        )
    ]
    .sort_values("review_id")
    .head(30)
)
```

```python
# ============================================
# SECTION 11 — DATA INTEGRATION
# 11.1 RELATIONSHIP KEY VALIDATION
# ============================================

print("=" * 70)
print("SECTION 11 — DATA INTEGRATION")
print("11.1 RELATIONSHIP KEY VALIDATION")
print("=" * 70)


# ============================================
# ORDERS → ORDER ITEMS
# ============================================

order_item_orders = (
    order_items_clean["order_id"]
    .isin(orders_clean["order_id"])
)

print("\nORDERS → ORDER_ITEMS")

print(
    "Order-item rows with valid order_id:",
    order_item_orders.sum()
)

print(
    "Order-item rows with missing order_id:",
    (~order_item_orders).sum()
)


# ============================================
# ORDERS → PAYMENTS
# ============================================

payment_orders = (
    payments_clean["order_id"]
    .isin(orders_clean["order_id"])
)

print("\nORDERS → PAYMENTS")

print(
    "Payment rows with valid order_id:",
    payment_orders.sum()
)

print(
    "Payment rows with missing order_id:",
    (~payment_orders).sum()
)


# ============================================
# ORDERS → REVIEWS
# ============================================

review_orders = (
    reviews_clean["order_id"]
    .isin(orders_clean["order_id"])
)

print("\nORDERS → REVIEWS")

print(
    "Review rows with valid order_id:",
    review_orders.sum()
)

print(
    "Review rows with missing order_id:",
    (~review_orders).sum()
)


# ============================================
# ORDER ITEMS → PRODUCTS
# ============================================

item_products = (
    order_items_clean["product_id"]
    .isin(products_clean["product_id"])
)

print("\nORDER_ITEMS → PRODUCTS")

print(
    "Order-item rows with valid product_id:",
    item_products.sum()
)

print(
    "Order-item rows with missing product_id:",
    (~item_products).sum()
)


# ============================================
# ORDER ITEMS → SELLERS
# ============================================

item_sellers = (
    order_items_clean["seller_id"]
    .isin(sellers_clean["seller_id"])
)

print("\nORDER_ITEMS → SELLERS")

print(
    "Order-item rows with valid seller_id:",
    item_sellers.sum()
)

print(
    "Order-item rows with missing seller_id:",
    (~item_sellers).sum()
)


# ============================================
# ORDERS → CUSTOMERS
# ============================================

order_customers = (
    orders_clean["customer_id"]
    .isin(customers_clean["customer_id"])
)

print("\nORDERS → CUSTOMERS")

print(
    "Orders with valid customer_id:",
    order_customers.sum()
)

print(
    "Orders with missing customer_id:",
    (~order_customers).sum()
)


print("\n" + "=" * 70)
print("RELATIONSHIP VALIDATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# 11.2 RELATIONSHIP CARDINALITY ANALYSIS
# ============================================

print("=" * 70)
print("11.2 RELATIONSHIP CARDINALITY ANALYSIS")
print("=" * 70)


# ============================================
# ORDERS → ORDER ITEMS
# ============================================

items_per_order = (
    order_items_clean
    .groupby("order_id")
    .size()
)

print("\nORDERS → ORDER_ITEMS")

print("Orders with order items:", items_per_order.shape[0])
print("Maximum items in one order:", items_per_order.max())
print("Average items per order:", round(items_per_order.mean(), 2))

print(
    "Orders with multiple items:",
    (items_per_order > 1).sum()
)


# ============================================
# ORDERS → PAYMENTS
# ============================================

payments_per_order = (
    payments_clean
    .groupby("order_id")
    .size()
)

print("\nORDERS → PAYMENTS")

print("Orders with payments:", payments_per_order.shape[0])
print("Maximum payments in one order:", payments_per_order.max())
print("Average payments per order:", round(payments_per_order.mean(), 2))

print(
    "Orders with multiple payments:",
    (payments_per_order > 1).sum()
)


# ============================================
# ORDERS → REVIEWS
# ============================================

reviews_per_order = (
    reviews_clean
    .groupby("order_id")
    .size()
)

print("\nORDERS → REVIEWS")

print("Orders with reviews:", reviews_per_order.shape[0])
print("Maximum reviews for one order:", reviews_per_order.max())
print("Average reviews per order:", round(reviews_per_order.mean(), 2))

print(
    "Orders with multiple review rows:",
    (reviews_per_order > 1).sum()
)


# ============================================
# ORDERS → CUSTOMERS
# ============================================

orders_per_customer = (
    orders_clean
    .groupby("customer_id")
    .size()
)

print("\nORDERS → CUSTOMERS")

print("Customers with orders:", orders_per_customer.shape[0])
print("Maximum orders per customer:", orders_per_customer.max())
print("Average orders per customer:", round(orders_per_customer.mean(), 2))

print(
    "Customers with multiple orders:",
    (orders_per_customer > 1).sum()
)


# ============================================
# PRODUCTS → ORDER ITEMS
# ============================================

items_per_product = (
    order_items_clean
    .groupby("product_id")
    .size()
)

print("\nPRODUCTS → ORDER_ITEMS")

print("Products sold:", items_per_product.shape[0])
print("Maximum sales records for one product:", items_per_product.max())
print("Average sales records per product:", round(items_per_product.mean(), 2))


# ============================================
# SELLERS → ORDER ITEMS
# ============================================

items_per_seller = (
    order_items_clean
    .groupby("seller_id")
    .size()
)

print("\nSELLERS → ORDER_ITEMS")

print("Sellers with sales:", items_per_seller.shape[0])
print("Maximum order-item records for one seller:", items_per_seller.max())
print("Average order-item records per seller:", round(items_per_seller.mean(), 2))


print("\n" + "=" * 70)
print("CARDINALITY ANALYSIS COMPLETED")
print("=" * 70)
```

```python
# ============================================
# 11.3 CUSTOMER IDENTITY VALIDATION
# ============================================

print("=" * 70)
print("11.3 CUSTOMER IDENTITY VALIDATION")
print("=" * 70)


# ============================================
# CUSTOMER_ID → CUSTOMER_UNIQUE_ID
# ============================================

customer_mapping = (
    customers_clean
    .groupby("customer_unique_id")
    .agg(
        customer_ids=("customer_id", "nunique"),
        total_orders=("customer_id", "size")
    )
    .sort_values("total_orders", ascending=False)
)

print("\nCUSTOMER UNIQUE ID ANALYSIS")

print(
    "Total customer_id records:",
    customers_clean["customer_id"].nunique()
)

print(
    "Unique customer_unique_id:",
    customers_clean["customer_unique_id"].nunique()
)

print(
    "Customers with multiple customer_id records:",
    (customer_mapping["customer_ids"] > 1).sum()
)

print(
    "Maximum customer_id records for one customer_unique_id:",
    customer_mapping["customer_ids"].max()
)

print(
    "Average customer_id records per customer_unique_id:",
    round(customer_mapping["customer_ids"].mean(), 2)
)


# ============================================
# REPEAT CUSTOMER ANALYSIS
# ============================================

repeat_customers = customer_mapping[
    customer_mapping["total_orders"] > 1
]

print("\nREPEAT CUSTOMER ANALYSIS")

print(
    "Customers with multiple orders:",
    len(repeat_customers)
)

print(
    "Maximum orders by one customer:",
    repeat_customers["total_orders"].max()
)


# ============================================
# TOP REPEAT CUSTOMERS
# ============================================

print("\nTOP 20 CUSTOMERS BY ORDER COUNT")

display(
    repeat_customers
    .head(20)
)


# ============================================
# CUSTOMER IDENTITY MAPPING
# ============================================

customer_identity = (
    customers_clean[
        [
            "customer_id",
            "customer_unique_id"
        ]
    ]
    .drop_duplicates()
)

print("\nCUSTOMER IDENTITY MAPPING")

print(
    "Unique customer_id → customer_unique_id mappings:",
    len(customer_identity)
)

print(
    "Duplicate customer_id mappings:",
    customer_identity["customer_id"].duplicated().sum()
)


print("\n" + "=" * 70)
print("CUSTOMER IDENTITY VALIDATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# 11.5 INTEGRATED MODEL TABLE CONSTRUCTION
# ============================================

print("=" * 70)
print("SECTION 11.5 — INTEGRATED MODEL TABLE CONSTRUCTION")
print("=" * 70)


# ============================================
# DIM CUSTOMER
# ============================================

dim_customer = (
    customers_clean[
        [
            "customer_unique_id",
            "customer_city",
            "customer_state"
        ]
    ]
    .drop_duplicates("customer_unique_id")
    .reset_index(drop=True)
)

print("\nDIM_CUSTOMER")
print("Rows:", len(dim_customer))
print(
    "Unique customers:",
    dim_customer["customer_unique_id"].nunique()
)


# ============================================
# DIM PRODUCT
# ============================================

dim_product = (
    products_clean[
        [
            "product_id",
            "product_category_name",
            "product_name_lenght",
            "product_description_lenght",
            "product_photos_qty",
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ]
    .drop_duplicates("product_id")
    .reset_index(drop=True)
)

print("\nDIM_PRODUCT")
print("Rows:", len(dim_product))
print(
    "Unique products:",
    dim_product["product_id"].nunique()
)


# ============================================
# DIM SELLER
# ============================================

dim_seller = (
    sellers_clean[
        [
            "seller_id",
            "seller_city",
            "seller_state"
        ]
    ]
    .drop_duplicates("seller_id")
    .reset_index(drop=True)
)

print("\nDIM_SELLER")
print("Rows:", len(dim_seller))
print(
    "Unique sellers:",
    dim_seller["seller_id"].nunique()
)


# ============================================
# DIM ORDER
# ============================================

dim_order = (
    orders_clean[
        [
            "order_id",
            "customer_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ]
    .drop_duplicates("order_id")
    .reset_index(drop=True)
)

print("\nDIM_ORDER")
print("Rows:", len(dim_order))
print(
    "Unique orders:",
    dim_order["order_id"].nunique()
)


# ============================================
# FACT ORDER ITEMS
# ============================================

fact_order_items = (
    order_items_clean[
        [
            "order_id",
            "order_item_id",
            "product_id",
            "seller_id",
            "shipping_limit_date",
            "price",
            "freight_value"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nFACT_ORDER_ITEMS")
print("Rows:", len(fact_order_items))


# ============================================
# FACT PAYMENTS
# ============================================

fact_payments = (
    payments_clean[
        [
            "order_id",
            "payment_sequential",
            "payment_type",
            "payment_installments",
            "payment_value"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nFACT_PAYMENTS")
print("Rows:", len(fact_payments))


# ============================================
# FACT REVIEWS
# ============================================

fact_reviews = (
    reviews_clean[
        [
            "review_id",
            "order_id",
            "review_score",
            "review_creation_date",
            "review_answer_timestamp",
            "review_response_time",
            "response_time_hours",
            "response_time_group",
            "review_category",
            "comment_length",
            "comment_word_count",
            "has_comment",
            "has_title"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nFACT_REVIEWS")
print("Rows:", len(fact_reviews))


# ============================================
# MODEL SUMMARY
# ============================================

model_summary = pd.DataFrame({
    "table": [
        "Dim_Customer",
        "Dim_Product",
        "Dim_Seller",
        "Dim_Order",
        "Fact_Order_Items",
        "Fact_Payments",
        "Fact_Reviews"
    ],
    "rows": [
        len(dim_customer),
        len(dim_product),
        len(dim_seller),
        len(dim_order),
        len(fact_order_items),
        len(fact_payments),
        len(fact_reviews)
    ]
})

print("\nMODEL SUMMARY")
display(model_summary)


print("=" * 70)
print("INTEGRATED MODEL TABLE CONSTRUCTION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# 11.6 FINAL INTEGRATION VALIDATION
# ============================================

print("=" * 70)
print("SECTION 11.6 — FINAL INTEGRATION VALIDATION")
print("=" * 70)


# ============================================
# 1. ROW COUNT VALIDATION
# ============================================

print("\n1. ROW COUNT VALIDATION")
print("-" * 70)

row_validation = pd.DataFrame({
    "table": [
        "Dim_Customer",
        "Dim_Product",
        "Dim_Seller",
        "Dim_Order",
        "Fact_Order_Items",
        "Fact_Payments",
        "Fact_Reviews"
    ],
    "model_rows": [
        len(dim_customer),
        len(dim_product),
        len(dim_seller),
        len(dim_order),
        len(fact_order_items),
        len(fact_payments),
        len(fact_reviews)
    ],
    "source_rows": [
        len(customers_clean),
        len(products_clean),
        len(sellers_clean),
        len(orders_clean),
        len(order_items_clean),
        len(payments_clean),
        len(reviews_clean)
    ]
})

row_validation["difference"] = (
    row_validation["model_rows"]
    - row_validation["source_rows"]
)

display(row_validation)


# ============================================
# 2. PRIMARY KEY VALIDATION
# ============================================

print("\n2. PRIMARY KEY VALIDATION")
print("-" * 70)

primary_key_checks = {
    "Dim_Customer": ("customer_id", dim_customer),
    "Dim_Product": ("product_id", dim_product),
    "Dim_Seller": ("seller_id", dim_seller),
    "Dim_Order": ("order_id", dim_order)
}

for table_name, (key, df) in primary_key_checks.items():

    print(
        f"{table_name} | {key}:",
        df[key].duplicated().sum(),
        "duplicates"
    )


# ============================================
# 3. FACT → DIMENSION KEY VALIDATION
# ============================================

print("\n3. FACT → DIMENSION KEY VALIDATION")
print("-" * 70)


# Order Items → Order

valid_order_items = fact_order_items[
    fact_order_items["order_id"].isin(
        dim_order["order_id"]
    )
]

print(
    "Fact_Order_Items → Dim_Order:",
    len(valid_order_items),
    "/",
    len(fact_order_items),
    "valid"
)


# Order Items → Product

valid_product_items = fact_order_items[
    fact_order_items["product_id"].isin(
        dim_product["product_id"]
    )
]

print(
    "Fact_Order_Items → Dim_Product:",
    len(valid_product_items),
    "/",
    len(fact_order_items),
    "valid"
)


# Order Items → Seller

valid_seller_items = fact_order_items[
    fact_order_items["seller_id"].isin(
        dim_seller["seller_id"]
    )
]

print(
    "Fact_Order_Items → Dim_Seller:",
    len(valid_seller_items),
    "/",
    len(fact_order_items),
    "valid"
)


# Payments → Order

valid_payments = fact_payments[
    fact_payments["order_id"].isin(
        dim_order["order_id"]
    )
]

print(
    "Fact_Payments → Dim_Order:",
    len(valid_payments),
    "/",
    len(fact_payments),
    "valid"
)


# Reviews → Order

valid_reviews = fact_reviews[
    fact_reviews["order_id"].isin(
        dim_order["order_id"]
    )
]

print(
    "Fact_Reviews → Dim_Order:",
    len(valid_reviews),
    "/",
    len(fact_reviews),
    "valid"
)


# Orders → Customer

valid_orders = dim_order[
    dim_order["customer_id"].isin(
        dim_customer["customer_id"]
    )
]

print(
    "Dim_Order → Dim_Customer:",
    len(valid_orders),
    "/",
    len(dim_order),
    "valid"
)


# ============================================
# 4. FOREIGN KEY ORPHAN CHECK
# ============================================

print("\n4. ORPHAN KEY CHECK")
print("-" * 70)

print(
    "Orphan order-item orders:",
    len(fact_order_items)
    - len(valid_order_items)
)

print(
    "Orphan order-item products:",
    len(fact_order_items)
    - len(valid_product_items)
)

print(
    "Orphan order-item sellers:",
    len(fact_order_items)
    - len(valid_seller_items)
)

print(
    "Orphan payment orders:",
    len(fact_payments)
    - len(valid_payments)
)

print(
    "Orphan review orders:",
    len(fact_reviews)
    - len(valid_reviews)
)

print(
    "Orphan customer orders:",
    len(dim_order)
    - len(valid_orders)
)


# ============================================
# 5. FACT TABLE DUPLICATE CHECK
# ============================================

print("\n5. FACT TABLE DUPLICATE CHECK")
print("-" * 70)

print(
    "Exact duplicate order-item rows:",
    fact_order_items.duplicated().sum()
)

print(
    "Exact duplicate payment rows:",
    fact_payments.duplicated().sum()
)

print(
    "Exact duplicate review rows:",
    fact_reviews.duplicated().sum()
)


# ============================================
# 6. CUSTOMER IDENTITY VALIDATION
# ============================================

print("\n6. CUSTOMER IDENTITY VALIDATION")
print("-" * 70)

print(
    "Customer IDs:",
    dim_customer["customer_id"].nunique()
)

print(
    "Unique Customers:",
    dim_customer["customer_unique_id"].nunique()
)

print(
    "Repeat Customers:",
    (
        dim_customer
        .groupby("customer_unique_id")
        ["customer_id"]
        .nunique()
        .gt(1)
        .sum()
    )
)


# ============================================
# 7. FINAL MODEL SUMMARY
# ============================================

print("\n7. FINAL MODEL SUMMARY")
print("-" * 70)

final_model_summary = pd.DataFrame({
    "table": [
        "Dim_Customer",
        "Dim_Product",
        "Dim_Seller",
        "Dim_Order",
        "Fact_Order_Items",
        "Fact_Payments",
        "Fact_Reviews"
    ],
    "rows": [
        len(dim_customer),
        len(dim_product),
        len(dim_seller),
        len(dim_order),
        len(fact_order_items),
        len(fact_payments),
        len(fact_reviews)
    ]
})

display(final_model_summary)


# ============================================
# 8. FINAL STATUS
# ============================================

print("\n" + "=" * 70)
print("FINAL INTEGRATION VALIDATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 12 — POWER BI DATA MODEL
# 12.1 DATE DIMENSION CREATION
# ============================================

print("=" * 70)
print("SECTION 12.1 — DATE DIMENSION CREATION")
print("=" * 70)


# ============================================
# DETERMINE DATE RANGE
# ============================================

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

all_dates = pd.concat([
    orders_clean[col]
    for col in date_columns
]).dropna()

min_date = all_dates.min().normalize()
max_date = all_dates.max().normalize()

print("\nDATE RANGE")
print("Minimum date:", min_date)
print("Maximum date:", max_date)


# ============================================
# CREATE DATE TABLE
# ============================================

date_range = pd.date_range(
    start=min_date,
    end=max_date,
    freq="D"
)

dim_date = pd.DataFrame({
    "date": date_range
})


# ============================================
# DATE ATTRIBUTES
# ============================================

dim_date["year"] = dim_date["date"].dt.year

dim_date["quarter"] = (
    "Q" +
    dim_date["date"].dt.quarter.astype(str)
)

dim_date["month_number"] = (
    dim_date["date"].dt.month
)

dim_date["month_name"] = (
    dim_date["date"].dt.month_name()
)

dim_date["month_short"] = (
    dim_date["date"].dt.strftime("%b")
)

dim_date["year_month"] = (
    dim_date["date"].dt.to_period("M").astype(str)
)

dim_date["week_number"] = (
    dim_date["date"].dt.isocalendar().week.astype(int)
)

dim_date["day"] = (
    dim_date["date"].dt.day
)

dim_date["day_name"] = (
    dim_date["date"].dt.day_name()
)

dim_date["day_of_week"] = (
    dim_date["date"].dt.dayofweek + 1
)

dim_date["is_weekend"] = (
    dim_date["day_of_week"] >= 6
)


# ============================================
# SORTING COLUMNS
# ============================================

dim_date["month_year_sort"] = (
    dim_date["year"] * 100
    + dim_date["month_number"]
)


# ============================================
# VALIDATION
# ============================================

print("\nDATE DIMENSION VALIDATION")

print(
    "Rows:",
    len(dim_date)
)

print(
    "Unique dates:",
    dim_date["date"].nunique()
)

print(
    "Duplicate dates:",
    dim_date["date"].duplicated().sum()
)

print(
    "Missing dates:",
    dim_date["date"].isna().sum()
)


# ============================================
# SAMPLE
# ============================================

print("\nDATE DIMENSION SAMPLE")

display(
    dim_date.head(10)
)


# ============================================
# FINAL STATUS
# ============================================

print("=" * 70)
print("12.1 DATE DIMENSION CREATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# 12.2 DATE DIMENSION RELATIONSHIP VALIDATION
# ============================================

print("=" * 70)
print("SECTION 12.2 — DATE DIMENSION RELATIONSHIP VALIDATION")
print("=" * 70)


# ============================================
# DATE COVERAGE VALIDATION
# ============================================

date_min = dim_date["date"].min()
date_max = dim_date["date"].max()

print("\nDATE DIMENSION RANGE")
print("Minimum date:", date_min)
print("Maximum date:", date_max)


# ============================================
# ORDER DATE COVERAGE
# ============================================

order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

print("\nORDER DATE COVERAGE")
print("-" * 70)

for col in order_date_columns:

    valid_dates = orders_clean[col].dropna()

    print(f"\n{col}")
    print("Minimum:", valid_dates.min())
    print("Maximum:", valid_dates.max())

    outside_range = (
        (valid_dates.dt.date < date_min.date()) |
        (valid_dates.dt.date > date_max.date())
    ).sum()

    print("Dates outside Dim_Date range:", outside_range)


# ============================================
# PURCHASE DATE KEY VALIDATION
# ============================================

orders_with_purchase_date = (
    orders_clean["order_purchase_timestamp"]
    .dropna()
    .dt.normalize()
)

date_keys = set(
    dim_date["date"]
    .dt.normalize()
)

valid_purchase_dates = orders_with_purchase_date.isin(date_keys).sum()

print("\nPURCHASE DATE KEY VALIDATION")
print("-" * 70)
print(
    "Purchase dates found in Dim_Date:",
    valid_purchase_dates
)
print(
    "Purchase dates missing from Dim_Date:",
    len(orders_with_purchase_date) - valid_purchase_dates
)


# ============================================
# DATE DIMENSION DUPLICATE VALIDATION
# ============================================

print("\nDATE DIMENSION KEY VALIDATION")
print("-" * 70)

print(
    "Duplicate date keys:",
    dim_date["date"].duplicated().sum()
)

print(
    "Total date rows:",
    len(dim_date)
)

print(
    "Unique date keys:",
    dim_date["date"].nunique()
)


# ============================================
# FINAL STATUS
# ============================================

print("\n" + "=" * 70)
print("DATE DIMENSION RELATIONSHIP VALIDATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 13.1 — FINAL MODEL TABLE VALIDATION
# ============================================

print("=" * 70)
print("SECTION 13.1 — FINAL MODEL TABLE VALIDATION")
print("=" * 70)


# ============================================================
# 0. REBUILD DIM_CUSTOMER CORRECTLY
# ============================================================

dim_customer = (
    customers_clean[
        [
            "customer_unique_id",
            "customer_city",
            "customer_state"
        ]
    ]
    .drop_duplicates(subset=["customer_unique_id"])
    .reset_index(drop=True)
)


# ============================================================
# 1. ADD CUSTOMER_UNIQUE_ID TO DIM_ORDER
# ============================================================

customer_mapping = (
    customers_clean[
        [
            "customer_id",
            "customer_unique_id"
        ]
    ]
    .drop_duplicates(subset=["customer_id"])
)


dim_order = (
    orders_clean[
        [
            "order_id",
            "customer_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ]
    .drop_duplicates(subset=["order_id"])
    .reset_index(drop=True)
)


# Merge customer_unique_id into Dim_Order

dim_order = dim_order.merge(
    customer_mapping,
    on="customer_id",
    how="left",
    validate="one_to_one"
)


# Create purchase date

dim_order["order_purchase_date"] = (
    dim_order["order_purchase_timestamp"]
    .dt.normalize()
)


# ============================================================
# 2. DIMENSION TABLE VALIDATION
# ============================================================

print("\n")
print("1. DIMENSION TABLE VALIDATION")
print("-" * 70)


dimension_checks = {

    "Dim_Customer": (
        dim_customer,
        "customer_unique_id"
    ),

    "Dim_Product": (
        dim_product,
        "product_id"
    ),

    "Dim_Seller": (
        dim_seller,
        "seller_id"
    ),

    "Dim_Order": (
        dim_order,
        "order_id"
    ),

    "Dim_Date": (
        dim_date,
        "date"
    )
}


dimension_results = []


for table_name, (df, key) in dimension_checks.items():

    rows = len(df)

    unique_keys = df[key].nunique()

    duplicate_keys = df[key].duplicated().sum()

    missing_keys = df[key].isna().sum()

    status = (
        "PASS"
        if duplicate_keys == 0 and missing_keys == 0
        else "FAIL"
    )

    dimension_results.append({
        "table": table_name,
        "rows": rows,
        "unique_keys": unique_keys,
        "duplicate_keys": duplicate_keys,
        "missing_keys": missing_keys,
        "status": status
    })


dimension_validation = pd.DataFrame(dimension_results)

display(dimension_validation)


# ============================================================
# 3. FACT TABLE ROW COUNT VALIDATION
# ============================================================

print("\n")
print("2. FACT TABLE VALIDATION")
print("-" * 70)


fact_results = []


for table_name, df in {
    "Fact_Order_Items": fact_order_items,
    "Fact_Payments": fact_payments,
    "Fact_Reviews": fact_reviews
}.items():

    fact_results.append({
        "table": table_name,
        "rows": len(df),
        "exact_duplicate_rows": df.duplicated().sum()
    })


fact_validation = pd.DataFrame(fact_results)

display(fact_validation)


# ============================================================
# 4. FOREIGN KEY VALIDATION
# ============================================================

print("\n")
print("3. FACT → DIMENSION KEY VALIDATION")
print("-" * 70)


# Order Items → Orders

order_items_orders_valid = (
    fact_order_items["order_id"]
    .isin(dim_order["order_id"])
    .sum()
)


# Order Items → Products

order_items_products_valid = (
    fact_order_items["product_id"]
    .isin(dim_product["product_id"])
    .sum()
)


# Order Items → Sellers

order_items_sellers_valid = (
    fact_order_items["seller_id"]
    .isin(dim_seller["seller_id"])
    .sum()
)


# Payments → Orders

payments_orders_valid = (
    fact_payments["order_id"]
    .isin(dim_order["order_id"])
    .sum()
)


# Reviews → Orders

reviews_orders_valid = (
    fact_reviews["order_id"]
    .isin(dim_order["order_id"])
    .sum()
)


# Orders → Customers

orders_customers_valid = (
    dim_order["customer_id"]
    .isin(customers_clean["customer_id"])
    .sum()
)


# Orders → Customer Unique ID

orders_unique_customers_valid = (
    dim_order["customer_unique_id"]
    .isin(dim_customer["customer_unique_id"])
    .sum()
)


# Orders → Date

orders_dates_valid = (
    dim_order["order_purchase_date"]
    .isin(dim_date["date"])
    .sum()
)


foreign_key_results = pd.DataFrame({

    "relationship": [
        "Fact_Order_Items → Dim_Order",
        "Fact_Order_Items → Dim_Product",
        "Fact_Order_Items → Dim_Seller",
        "Fact_Payments → Dim_Order",
        "Fact_Reviews → Dim_Order",
        "Dim_Order → Customer",
        "Dim_Order → Dim_Customer",
        "Dim_Order → Dim_Date"
    ],

    "valid_rows": [
        order_items_orders_valid,
        order_items_products_valid,
        order_items_sellers_valid,
        payments_orders_valid,
        reviews_orders_valid,
        orders_customers_valid,
        orders_unique_customers_valid,
        orders_dates_valid
    ],

    "total_rows": [
        len(fact_order_items),
        len(fact_order_items),
        len(fact_order_items),
        len(fact_payments),
        len(fact_reviews),
        len(dim_order),
        len(dim_order),
        len(dim_order)
    ]
})


foreign_key_results["invalid_rows"] = (
    foreign_key_results["total_rows"]
    - foreign_key_results["valid_rows"]
)


foreign_key_results["status"] = (
    foreign_key_results["invalid_rows"]
    .eq(0)
    .map({
        True: "PASS",
        False: "FAIL"
    })
)


display(foreign_key_results)


# ============================================================
# 5. ORPHAN KEY VALIDATION
# ============================================================

print("\n")
print("4. ORPHAN KEY VALIDATION")
print("-" * 70)


orphan_results = {

    "Order-item orders": (
        ~fact_order_items["order_id"]
        .isin(dim_order["order_id"])
    ).sum(),

    "Order-item products": (
        ~fact_order_items["product_id"]
        .isin(dim_product["product_id"])
    ).sum(),

    "Order-item sellers": (
        ~fact_order_items["seller_id"]
        .isin(dim_seller["seller_id"])
    ).sum(),

    "Payment orders": (
        ~fact_payments["order_id"]
        .isin(dim_order["order_id"])
    ).sum(),

    "Review orders": (
        ~fact_reviews["order_id"]
        .isin(dim_order["order_id"])
    ).sum(),

    "Customer orders": (
        ~dim_order["customer_unique_id"]
        .isin(dim_customer["customer_unique_id"])
    ).sum(),

    "Order dates": (
        ~dim_order["order_purchase_date"]
        .isin(dim_date["date"])
    ).sum()
}


for relationship, orphan_count in orphan_results.items():

    print(
        f"{relationship:25} : "
        f"{orphan_count}"
    )


# ============================================================
# 6. CRITICAL NULL VALIDATION
# ============================================================

print("\n")
print("5. CRITICAL NULL VALIDATION")
print("-" * 70)


critical_columns = {

    "Dim_Customer": (
        dim_customer,
        [
            "customer_unique_id"
        ]
    ),

    "Dim_Product": (
        dim_product,
        [
            "product_id"
        ]
    ),

    "Dim_Seller": (
        dim_seller,
        [
            "seller_id"
        ]
    ),

    "Dim_Order": (
        dim_order,
        [
            "order_id",
            "customer_id",
            "customer_unique_id",
            "order_purchase_date"
        ]
    ),

    "Dim_Date": (
        dim_date,
        [
            "date"
        ]
    ),

    "Fact_Order_Items": (
        fact_order_items,
        [
            "order_id",
            "product_id",
            "seller_id"
        ]
    ),

    "Fact_Payments": (
        fact_payments,
        [
            "order_id"
        ]
    ),

    "Fact_Reviews": (
        fact_reviews,
        [
            "review_id",
            "order_id"
        ]
    )
}


null_results = []


for table_name, (df, columns) in critical_columns.items():

    for column in columns:

        missing_count = df[column].isna().sum()

        null_results.append({

            "table": table_name,

            "column": column,

            "missing_count": missing_count,

            "status": (
                "PASS"
                if missing_count == 0
                else "CHECK"
            )
        })


null_validation = pd.DataFrame(null_results)

display(null_validation)


# ============================================================
# 7. EXACT DUPLICATE ROW VALIDATION
# ============================================================

print("\n")
print("6. EXACT DUPLICATE ROW VALIDATION")
print("-" * 70)


duplicate_results = []


for table_name, df in {

    "Dim_Customer": dim_customer,

    "Dim_Product": dim_product,

    "Dim_Seller": dim_seller,

    "Dim_Order": dim_order,

    "Dim_Date": dim_date,

    "Fact_Order_Items": fact_order_items,

    "Fact_Payments": fact_payments,

    "Fact_Reviews": fact_reviews

}.items():

    duplicate_count = df.duplicated().sum()

    duplicate_results.append({

        "table": table_name,

        "exact_duplicate_rows": duplicate_count,

        "status": (
            "PASS"
            if duplicate_count == 0
            else "CHECK"
        )
    })


duplicate_validation = pd.DataFrame(duplicate_results)

display(duplicate_validation)


# ============================================================
# 8. CUSTOMER IDENTITY VALIDATION
# ============================================================

print("\n")
print("7. CUSTOMER IDENTITY VALIDATION")
print("-" * 70)


total_customer_ids = (
    customers_clean["customer_id"]
    .nunique()
)


unique_customers = (
    customers_clean["customer_unique_id"]
    .nunique()
)


repeat_customer_count = (
    customers_clean
    .groupby("customer_unique_id")
    ["customer_id"]
    .nunique()
    .gt(1)
    .sum()
)


print(
    "Customer IDs:",
    total_customer_ids
)


print(
    "Unique Customers:",
    unique_customers
)


print(
    "Repeat Customers:",
    repeat_customer_count
)


print(
    "Dim_Customer rows:",
    len(dim_customer)
)


print(
    "Dim_Customer unique customer_unique_id:",
    dim_customer["customer_unique_id"].nunique()
)


# ============================================================
# 9. DIMENSION ROW COUNT SUMMARY
# ============================================================

print("\n")
print("8. FINAL MODEL ROW COUNTS")
print("-" * 70)


model_summary = pd.DataFrame({

    "table": [

        "Dim_Customer",
        "Dim_Product",
        "Dim_Seller",
        "Dim_Order",
        "Dim_Date",
        "Fact_Order_Items",
        "Fact_Payments",
        "Fact_Reviews"

    ],

    "rows": [

        len(dim_customer),
        len(dim_product),
        len(dim_seller),
        len(dim_order),
        len(dim_date),
        len(fact_order_items),
        len(fact_payments),
        len(fact_reviews)

    ]
})


display(model_summary)


# ============================================================
# 10. FINAL PASS / FAIL CHECK
# ============================================================

print("\n")
print("=" * 70)
print("FINAL MODEL VALIDATION STATUS")
print("=" * 70)


dimension_pass = (
    dimension_validation["status"]
    .eq("PASS")
    .all()
)


foreign_key_pass = (
    foreign_key_results["status"]
    .eq("PASS")
    .all()
)


orphan_pass = (
    all(
        count == 0
        for count in orphan_results.values()
    )
)


critical_null_pass = (
    null_validation["status"]
    .eq("PASS")
    .all()
)


duplicate_pass = (
    duplicate_validation["status"]
    .eq("PASS")
    .all()
)


customer_dimension_pass = (
    len(dim_customer) == 96096
    and
    dim_customer["customer_unique_id"].nunique() == 96096
    and
    dim_customer["customer_unique_id"].duplicated().sum() == 0
)


if all([
    dimension_pass,
    foreign_key_pass,
    orphan_pass,
    critical_null_pass,
    duplicate_pass,
    customer_dimension_pass
]):

    print("STATUS: PASS")

    print(
        "All model validation checks completed successfully."
    )

else:

    print("STATUS: CHECK")

    print(
        "One or more validation checks require review."
    )


print("=" * 70)
print("SECTION 13.1 VALIDATION COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 13.2 — POWER BI-READY MODEL PREPARATION
# ============================================

print("=" * 70)
print("SECTION 13.2 — POWER BI-READY MODEL PREPARATION")
print("=" * 70)


# ============================================================
# 1. FINAL COPY OF DIM_CUSTOMER
# ============================================================

powerbi_dim_customer = dim_customer.copy()

print("\nDIM_CUSTOMER")
print("Rows:", len(powerbi_dim_customer))
print(
    "Unique customers:",
    powerbi_dim_customer["customer_unique_id"].nunique()
)


# ============================================================
# 2. FINAL COPY OF DIM_PRODUCT
# ============================================================

powerbi_dim_product = dim_product.copy()

print("\nDIM_PRODUCT")
print("Rows:", len(powerbi_dim_product))
print(
    "Unique products:",
    powerbi_dim_product["product_id"].nunique()
)


# ============================================================
# 3. FINAL COPY OF DIM_SELLER
# ============================================================

powerbi_dim_seller = dim_seller.copy()

print("\nDIM_SELLER")
print("Rows:", len(powerbi_dim_seller))
print(
    "Unique sellers:",
    powerbi_dim_seller["seller_id"].nunique()
)


# ============================================================
# 4. FINAL COPY OF DIM_ORDER
# ============================================================

powerbi_dim_order = dim_order.copy()

print("\nDIM_ORDER")
print("Rows:", len(powerbi_dim_order))
print(
    "Unique orders:",
    powerbi_dim_order["order_id"].nunique()
)


# ============================================================
# 5. FINAL COPY OF DIM_DATE
# ============================================================

powerbi_dim_date = dim_date.copy()

print("\nDIM_DATE")
print("Rows:", len(powerbi_dim_date))
print(
    "Unique dates:",
    powerbi_dim_date["date"].nunique()
)


# ============================================================
# 6. FINAL COPY OF FACT_ORDER_ITEMS
# ============================================================

powerbi_fact_order_items = fact_order_items.copy()

print("\nFACT_ORDER_ITEMS")
print("Rows:", len(powerbi_fact_order_items))


# ============================================================
# 7. FINAL COPY OF FACT_PAYMENTS
# ============================================================

powerbi_fact_payments = fact_payments.copy()

print("\nFACT_PAYMENTS")
print("Rows:", len(powerbi_fact_payments))


# ============================================================
# 8. FINAL COPY OF FACT_REVIEWS
# ============================================================

powerbi_fact_reviews = fact_reviews.copy()

print("\nFACT_REVIEWS")
print("Rows:", len(powerbi_fact_reviews))


# ============================================================
# 9. DATA TYPE VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("DATA TYPE VALIDATION")
print("=" * 70)


# Date columns

powerbi_dim_order["order_purchase_timestamp"] = pd.to_datetime(
    powerbi_dim_order["order_purchase_timestamp"],
    errors="coerce"
)

powerbi_dim_order["order_approved_at"] = pd.to_datetime(
    powerbi_dim_order["order_approved_at"],
    errors="coerce"
)

powerbi_dim_order["order_delivered_carrier_date"] = pd.to_datetime(
    powerbi_dim_order["order_delivered_carrier_date"],
    errors="coerce"
)

powerbi_dim_order["order_delivered_customer_date"] = pd.to_datetime(
    powerbi_dim_order["order_delivered_customer_date"],
    errors="coerce"
)

powerbi_dim_order["order_estimated_delivery_date"] = pd.to_datetime(
    powerbi_dim_order["order_estimated_delivery_date"],
    errors="coerce"
)

powerbi_dim_order["order_purchase_date"] = pd.to_datetime(
    powerbi_dim_order["order_purchase_date"],
    errors="coerce"
)

powerbi_dim_date["date"] = pd.to_datetime(
    powerbi_dim_date["date"],
    errors="coerce"
)

powerbi_fact_order_items["shipping_limit_date"] = pd.to_datetime(
    powerbi_fact_order_items["shipping_limit_date"],
    errors="coerce"
)

powerbi_fact_reviews["review_creation_date"] = pd.to_datetime(
    powerbi_fact_reviews["review_creation_date"],
    errors="coerce"
)

powerbi_fact_reviews["review_answer_timestamp"] = pd.to_datetime(
    powerbi_fact_reviews["review_answer_timestamp"],
    errors="coerce"
)


print("\nDim_Order date columns:")
print(
    powerbi_dim_order[
        [
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date",
            "order_purchase_date"
        ]
    ].dtypes
)


# ============================================================
# 10. CREATE BUSINESS-FRIENDLY DATE COLUMNS
# ============================================================

print("\n")
print("=" * 70)
print("CREATING BUSINESS-FRIENDLY DATE COLUMNS")
print("=" * 70)


powerbi_dim_date["year"] = (
    powerbi_dim_date["year"]
    .astype("int64")
)

powerbi_dim_date["month_number"] = (
    powerbi_dim_date["month_number"]
    .astype("int64")
)

powerbi_dim_date["day"] = (
    powerbi_dim_date["day"]
    .astype("int64")
)

powerbi_dim_date["day_of_week"] = (
    powerbi_dim_date["day_of_week"]
    .astype("int64")
)

powerbi_dim_date["is_weekend"] = (
    powerbi_dim_date["is_weekend"]
    .astype(bool)
)


# ============================================================
# 11. CREATE SORT COLUMNS FOR POWER BI
# ============================================================

powerbi_dim_date["month_year_sort"] = (
    powerbi_dim_date["year"] * 100
    + powerbi_dim_date["month_number"]
)


# ============================================================
# 12. FINAL TABLE SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("FINAL POWER BI TABLE SUMMARY")
print("=" * 70)


powerbi_model_summary = pd.DataFrame({

    "table": [

        "Dim_Customer",
        "Dim_Product",
        "Dim_Seller",
        "Dim_Order",
        "Dim_Date",
        "Fact_Order_Items",
        "Fact_Payments",
        "Fact_Reviews"

    ],

    "rows": [

        len(powerbi_dim_customer),
        len(powerbi_dim_product),
        len(powerbi_dim_seller),
        len(powerbi_dim_order),
        len(powerbi_dim_date),
        len(powerbi_fact_order_items),
        len(powerbi_fact_payments),
        len(powerbi_fact_reviews)

    ],

    "primary_key": [

        "customer_unique_id",
        "product_id",
        "seller_id",
        "order_id",
        "date",
        "order_id + order_item_id",
        "order_id + payment_sequential",
        "review_id"

    ]

})


display(powerbi_model_summary)


# ============================================================
# 13. FINAL PRIMARY KEY CHECK
# ============================================================

print("\n")
print("=" * 70)
print("FINAL PRIMARY KEY CHECK")
print("=" * 70)


final_key_checks = {

    "Dim_Customer": (
        powerbi_dim_customer["customer_unique_id"]
        .duplicated()
        .sum()
    ),

    "Dim_Product": (
        powerbi_dim_product["product_id"]
        .duplicated()
        .sum()
    ),

    "Dim_Seller": (
        powerbi_dim_seller["seller_id"]
        .duplicated()
        .sum()
    ),

    "Dim_Order": (
        powerbi_dim_order["order_id"]
        .duplicated()
        .sum()
    ),

    "Dim_Date": (
        powerbi_dim_date["date"]
        .duplicated()
        .sum()
    )
}


for table_name, duplicate_count in final_key_checks.items():

    print(
        f"{table_name:20} : "
        f"{duplicate_count} duplicate keys"
    )


# ============================================================
# 14. FINAL NULL CHECK FOR KEYS
# ============================================================

print("\n")
print("=" * 70)
print("FINAL KEY NULL CHECK")
print("=" * 70)


final_key_nulls = {

    "Dim_Customer": (
        powerbi_dim_customer["customer_unique_id"]
        .isna()
        .sum()
    ),

    "Dim_Product": (
        powerbi_dim_product["product_id"]
        .isna()
        .sum()
    ),

    "Dim_Seller": (
        powerbi_dim_seller["seller_id"]
        .isna()
        .sum()
    ),

    "Dim_Order": (
        powerbi_dim_order["order_id"]
        .isna()
        .sum()
    ),

    "Dim_Date": (
        powerbi_dim_date["date"]
        .isna()
        .sum()
    )

}


for table_name, missing_count in final_key_nulls.items():

    print(
        f"{table_name:20} : "
        f"{missing_count} missing keys"
    )


# ============================================================
# 15. FINAL ROW COUNT VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("FINAL ROW COUNT VALIDATION")
print("=" * 70)


expected_rows = {

    "Dim_Customer": 96096,
    "Dim_Product": 32951,
    "Dim_Seller": 3095,
    "Dim_Order": 99441,
    "Dim_Date": 800,
    "Fact_Order_Items": 112650,
    "Fact_Payments": 103886,
    "Fact_Reviews": 99224
}


actual_rows = {

    "Dim_Customer": len(powerbi_dim_customer),
    "Dim_Product": len(powerbi_dim_product),
    "Dim_Seller": len(powerbi_dim_seller),
    "Dim_Order": len(powerbi_dim_order),
    "Dim_Date": len(powerbi_dim_date),
    "Fact_Order_Items": len(powerbi_fact_order_items),
    "Fact_Payments": len(powerbi_fact_payments),
    "Fact_Reviews": len(powerbi_fact_reviews)
}


row_validation = []


for table_name in expected_rows:

    expected = expected_rows[table_name]

    actual = actual_rows[table_name]

    difference = actual - expected

    row_validation.append({

        "table": table_name,

        "expected_rows": expected,

        "actual_rows": actual,

        "difference": difference,

        "status": (
            "PASS"
            if difference == 0
            else "FAIL"
        )

    })


row_validation = pd.DataFrame(row_validation)

display(row_validation)


# ============================================================
# 16. FINAL STATUS
# ============================================================

print("\n")
print("=" * 70)
print("SECTION 13.2 FINAL STATUS")
print("=" * 70)


keys_pass = all(
    count == 0
    for count in final_key_checks.values()
)


nulls_pass = all(
    count == 0
    for count in final_key_nulls.values()
)


rows_pass = (
    row_validation["status"]
    .eq("PASS")
    .all()
)


if keys_pass and nulls_pass and rows_pass:

    print("STATUS: PASS")
    print(
        "All Power BI-ready tables prepared successfully."
    )

else:

    print("STATUS: CHECK")
    print(
        "One or more Power BI preparation checks require review."
    )


print("=" * 70)
print("SECTION 13.2 COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 14.1 — OVERALL BUSINESS KPIs
# ============================================

print("=" * 70)
print("SECTION 14.1 — OVERALL BUSINESS KPIs")
print("=" * 70)


# ============================================
# 1. ORDER KPIs
# ============================================

total_orders = dim_order["order_id"].nunique()

total_customers = dim_customer["customer_unique_id"].nunique()

total_customer_ids = dim_order["customer_id"].nunique()

total_products = dim_product["product_id"].nunique()

total_sellers = dim_seller["seller_id"].nunique()


# ============================================
# 2. ORDER ITEM KPIs
# ============================================

total_order_items = len(fact_order_items)

total_items_sold = fact_order_items["order_item_id"].count()

total_revenue = fact_order_items["price"].sum()

total_freight = fact_order_items["freight_value"].sum()

total_sales_value = total_revenue + total_freight


# ============================================
# 3. AVERAGE KPIs
# ============================================

average_order_value = (
    total_revenue / total_orders
)

average_item_price = (
    fact_order_items["price"].mean()
)

average_freight_per_item = (
    fact_order_items["freight_value"].mean()
)

average_items_per_order = (
    total_order_items / total_orders
)


# ============================================
# 4. PAYMENT KPIs
# ============================================

total_payment_records = len(fact_payments)

total_payment_value = (
    fact_payments["payment_value"].sum()
)

average_payment_value = (
    fact_payments["payment_value"].mean()
)


# ============================================
# 5. REVIEW KPIs
# ============================================

total_reviews = len(fact_reviews)

average_review_score = (
    fact_reviews["review_score"].mean()
)

positive_reviews = (
    (fact_reviews["review_score"] >= 4).sum()
)

negative_reviews = (
    (fact_reviews["review_score"] <= 2).sum()
)

neutral_reviews = (
    (fact_reviews["review_score"] == 3).sum()
)


# ============================================
# 6. REVIEW PERCENTAGES
# ============================================

positive_review_pct = (
    positive_reviews / total_reviews * 100
)

negative_review_pct = (
    negative_reviews / total_reviews * 100
)

neutral_review_pct = (
    neutral_reviews / total_reviews * 100
)


# ============================================
# 7. CUSTOMER KPIs
# ============================================

customer_order_counts = (
    dim_order
    .merge(
        dim_customer[
            ["customer_unique_id"]
        ],
        on="customer_unique_id",
        how="left"
    )
    .groupby("customer_unique_id")["order_id"]
    .count()
)

repeat_customers = (
    (customer_order_counts > 1).sum()
)

one_time_customers = (
    (customer_order_counts == 1).sum()
)

repeat_customer_rate = (
    repeat_customers /
    total_customers * 100
)


# ============================================
# 8. KPI SUMMARY
# ============================================

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Orders",
        "Unique Customers",
        "Customer IDs",
        "Total Products",
        "Total Sellers",
        "Total Order Items",
        "Total Revenue",
        "Total Freight",
        "Total Sales + Freight",
        "Average Order Value",
        "Average Item Price",
        "Average Freight per Item",
        "Average Items per Order",
        "Total Payment Records",
        "Total Payment Value",
        "Average Payment Value",
        "Total Reviews",
        "Average Review Score",
        "Positive Reviews",
        "Neutral Reviews",
        "Negative Reviews",
        "Positive Review %",
        "Neutral Review %",
        "Negative Review %",
        "Repeat Customers",
        "One-Time Customers",
        "Repeat Customer Rate %"
    ],
    
    "Value": [
        total_orders,
        total_customers,
        total_customer_ids,
        total_products,
        total_sellers,
        total_order_items,
        total_revenue,
        total_freight,
        total_sales_value,
        average_order_value,
        average_item_price,
        average_freight_per_item,
        average_items_per_order,
        total_payment_records,
        total_payment_value,
        average_payment_value,
        total_reviews,
        average_review_score,
        positive_reviews,
        neutral_reviews,
        negative_reviews,
        positive_review_pct,
        neutral_review_pct,
        negative_review_pct,
        repeat_customers,
        one_time_customers,
        repeat_customer_rate
    ]
})


# ============================================
# 9. DISPLAY KPI SUMMARY
# ============================================

display(kpi_summary)


# ============================================
# 10. FORMATTED BUSINESS SUMMARY
# ============================================

print("\n" + "=" * 70)
print("BUSINESS KPI SUMMARY")
print("=" * 70)

print(f"Total Orders              : {total_orders:,}")
print(f"Unique Customers          : {total_customers:,}")
print(f"Customer IDs              : {total_customer_ids:,}")
print(f"Total Products            : {total_products:,}")
print(f"Total Sellers             : {total_sellers:,}")
print(f"Total Order Items         : {total_order_items:,}")

print(f"\nTotal Revenue             : {total_revenue:,.2f}")
print(f"Total Freight             : {total_freight:,.2f}")
print(f"Total Sales + Freight     : {total_sales_value:,.2f}")

print(f"\nAverage Order Value       : {average_order_value:,.2f}")
print(f"Average Item Price        : {average_item_price:,.2f}")
print(f"Average Freight / Item    : {average_freight_per_item:,.2f}")
print(f"Average Items / Order     : {average_items_per_order:.2f}")

print(f"\nPayment Records            : {total_payment_records:,}")
print(f"Payment Value              : {total_payment_value:,.2f}")
print(f"Average Payment Value      : {average_payment_value:,.2f}")

print(f"\nTotal Reviews              : {total_reviews:,}")
print(f"Average Review Score       : {average_review_score:.2f}")

print(f"\nPositive Reviews           : {positive_reviews:,}")
print(f"Neutral Reviews            : {neutral_reviews:,}")
print(f"Negative Reviews           : {negative_reviews:,}")

print(f"\nPositive Review %          : {positive_review_pct:.2f}%")
print(f"Neutral Review %           : {neutral_review_pct:.2f}%")
print(f"Negative Review %          : {negative_review_pct:.2f}%")

print(f"\nRepeat Customers           : {repeat_customers:,}")
print(f"One-Time Customers         : {one_time_customers:,}")
print(f"Repeat Customer Rate       : {repeat_customer_rate:.2f}%")


print("\n" + "=" * 70)
print("SECTION 14.1 COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 14.2 — SALES & REVENUE ANALYSIS
# ============================================

print("=" * 70)
print("SECTION 14.2 — SALES & REVENUE ANALYSIS")
print("=" * 70)


# ============================================
# 1. PREPARE SALES DATA
# ============================================

sales_data = (
    fact_order_items
    .merge(
        dim_order[
            [
                "order_id",
                "order_status",
                "order_purchase_timestamp"
            ]
        ],
        on="order_id",
        how="left"
    )
)

sales_data["order_purchase_timestamp"] = pd.to_datetime(
    sales_data["order_purchase_timestamp"],
    errors="coerce"
)

sales_data["year"] = (
    sales_data["order_purchase_timestamp"].dt.year
)

sales_data["month_number"] = (
    sales_data["order_purchase_timestamp"].dt.month
)

sales_data["month_name"] = (
    sales_data["order_purchase_timestamp"].dt.strftime("%B")
)

sales_data["year_month"] = (
    sales_data["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)


# ============================================
# 2. OVERALL SALES SUMMARY
# ============================================

sales_summary = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Order Items",
        "Total Revenue",
        "Total Freight",
        "Total Sales + Freight",
        "Average Order Value",
        "Average Item Price",
        "Average Freight per Item"
    ],
    
    "Value": [
        sales_data["order_id"].nunique(),
        len(sales_data),
        sales_data["price"].sum(),
        sales_data["freight_value"].sum(),
        (
            sales_data["price"].sum()
            +
            sales_data["freight_value"].sum()
        ),
        (
            sales_data["price"].sum()
            /
            sales_data["order_id"].nunique()
        ),
        sales_data["price"].mean(),
        sales_data["freight_value"].mean()
    ]
})

print("\nOVERALL SALES SUMMARY")
display(sales_summary)


# ============================================
# 3. MONTHLY SALES ANALYSIS
# ============================================

monthly_sales = (
    sales_data
    .groupby("year_month")
    .agg(
        orders=("order_id", "nunique"),
        order_items=("order_item_id", "count"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

monthly_sales["total_sales"] = (
    monthly_sales["revenue"]
    +
    monthly_sales["freight"]
)

monthly_sales["average_order_value"] = (
    monthly_sales["revenue"]
    /
    monthly_sales["orders"]
)

monthly_sales["revenue_growth_pct"] = (
    monthly_sales["revenue"]
    .pct_change()
    *
    100
)

print("\nMONTHLY SALES ANALYSIS")
display(monthly_sales)


# ============================================
# 4. TOP 10 MONTHS BY REVENUE
# ============================================

top_revenue_months = (
    monthly_sales
    .sort_values(
        "revenue",
        ascending=False
    )
    .head(10)
    .reset_index(drop=True)
)

print("\nTOP 10 MONTHS BY REVENUE")
display(top_revenue_months)


# ============================================
# 5. LOWEST 10 MONTHS BY REVENUE
# ============================================

lowest_revenue_months = (
    monthly_sales
    .sort_values(
        "revenue",
        ascending=True
    )
    .head(10)
    .reset_index(drop=True)
)

print("\nLOWEST 10 MONTHS BY REVENUE")
display(lowest_revenue_months)


# ============================================
# 6. YEARLY SALES ANALYSIS
# ============================================

yearly_sales = (
    sales_data
    .groupby("year")
    .agg(
        orders=("order_id", "nunique"),
        order_items=("order_item_id", "count"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

yearly_sales["total_sales"] = (
    yearly_sales["revenue"]
    +
    yearly_sales["freight"]
)

yearly_sales["average_order_value"] = (
    yearly_sales["revenue"]
    /
    yearly_sales["orders"]
)

yearly_sales["revenue_growth_pct"] = (
    yearly_sales["revenue"]
    .pct_change()
    *
    100
)

print("\nYEARLY SALES ANALYSIS")
display(yearly_sales)


# ============================================
# 7. YEAR-MONTH SALES TREND
# ============================================

monthly_sales_trend = (
    monthly_sales[
        [
            "year_month",
            "orders",
            "revenue",
            "freight",
            "total_sales",
            "average_order_value"
        ]
    ]
    .copy()
)

print("\nMONTHLY SALES TREND")
display(monthly_sales_trend)


# ============================================
# 8. SALES CONCENTRATION
# ============================================

monthly_sales_sorted = (
    monthly_sales
    .sort_values(
        "revenue",
        ascending=False
    )
    .reset_index(drop=True)
)

monthly_sales_sorted["revenue_share_pct"] = (
    monthly_sales_sorted["revenue"]
    /
    monthly_sales_sorted["revenue"].sum()
    *
    100
)

monthly_sales_sorted["cumulative_revenue_pct"] = (
    monthly_sales_sorted["revenue_share_pct"]
    .cumsum()
)

print("\nREVENUE CONCENTRATION")
display(
    monthly_sales_sorted[
        [
            "year_month",
            "revenue",
            "revenue_share_pct",
            "cumulative_revenue_pct"
        ]
    ]
)


# ============================================
# 9. PEAK SALES PERIOD
# ============================================

peak_month = (
    monthly_sales
    .loc[
        monthly_sales["revenue"].idxmax()
    ]
)

lowest_month = (
    monthly_sales
    .loc[
        monthly_sales["revenue"].idxmin()
    ]
)

print("\n" + "=" * 70)
print("SALES PERFORMANCE HIGHLIGHTS")
print("=" * 70)

print(
    f"Peak Revenue Month       : "
    f"{peak_month['year_month']}"
)

print(
    f"Peak Month Revenue       : "
    f"{peak_month['revenue']:,.2f}"
)

print(
    f"Peak Month Orders        : "
    f"{peak_month['orders']:,}"
)

print(
    f"Lowest Revenue Month     : "
    f"{lowest_month['year_month']}"
)

print(
    f"Lowest Month Revenue     : "
    f"{lowest_month['revenue']:,.2f}"
)

print(
    f"Lowest Month Orders      : "
    f"{lowest_month['orders']:,}"
)


# ============================================
# 10. FINAL STATUS
# ============================================

print("\n" + "=" * 70)
print("SECTION 14.2 COMPLETED")
print("=" * 70)
```

```python
# ============================================
# SECTION 14.3 — PRODUCT & CATEGORY PERFORMANCE
# ============================================

print("=" * 70)
print("SECTION 14.3 — PRODUCT & CATEGORY PERFORMANCE")
print("=" * 70)


# ============================================
# 1. PREPARE PRODUCT SALES DATA
# ============================================

product_sales = (
    fact_order_items
    .merge(
        dim_product[
            [
                "product_id",
                "product_category_name"
            ]
        ],
        on="product_id",
        how="left"
    )
)

product_sales["product_category_name"] = (
    product_sales["product_category_name"]
    .fillna("Unknown")
)


# ============================================
# 2. OVERALL PRODUCT PERFORMANCE
# ============================================

product_summary = (
    product_sales
    .groupby("product_id")
    .agg(
        order_items=("order_item_id", "count"),
        unique_orders=("order_id", "nunique"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

product_summary["total_sales"] = (
    product_summary["revenue"]
    +
    product_summary["freight"]
)

product_summary["average_price"] = (
    product_summary["revenue"]
    /
    product_summary["order_items"]
)


print("\nPRODUCT PERFORMANCE SUMMARY")

display(
    product_summary.describe(
        include="all"
    )
)


# ============================================
# 3. TOP 20 PRODUCTS BY REVENUE
# ============================================

top_products_revenue = (
    product_summary
    .sort_values(
        "revenue",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print("\nTOP 20 PRODUCTS BY REVENUE")

display(top_products_revenue)


# ============================================
# 4. TOP 20 PRODUCTS BY SALES VOLUME
# ============================================

top_products_volume = (
    product_summary
    .sort_values(
        "order_items",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print("\nTOP 20 PRODUCTS BY SALES VOLUME")

display(top_products_volume)


# ============================================
# 5. PRODUCT CATEGORY PERFORMANCE
# ============================================

category_performance = (
    product_sales
    .groupby("product_category_name")
    .agg(
        order_items=("order_item_id", "count"),
        unique_orders=("order_id", "nunique"),
        unique_products=("product_id", "nunique"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

category_performance["total_sales"] = (
    category_performance["revenue"]
    +
    category_performance["freight"]
)

category_performance["average_item_price"] = (
    category_performance["revenue"]
    /
    category_performance["order_items"]
)

category_performance["revenue_share_pct"] = (
    category_performance["revenue"]
    /
    category_performance["revenue"].sum()
    *
    100
)


# ============================================
# 6. TOP 20 CATEGORIES BY REVENUE
# ============================================

top_categories_revenue = (
    category_performance
    .sort_values(
        "revenue",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print("\nTOP 20 PRODUCT CATEGORIES BY REVENUE")

display(top_categories_revenue)


# ============================================
# 7. TOP 20 CATEGORIES BY ORDER ITEMS
# ============================================

top_categories_volume = (
    category_performance
    .sort_values(
        "order_items",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print("\nTOP 20 PRODUCT CATEGORIES BY SALES VOLUME")

display(top_categories_volume)


# ============================================
# 8. LOWEST PERFORMING CATEGORIES
# ============================================

lowest_categories = (
    category_performance
    .sort_values(
        "revenue",
        ascending=True
    )
    .head(10)
    .reset_index(drop=True)
)

print("\nLOWEST 10 CATEGORIES BY REVENUE")

display(lowest_categories)


# ============================================
# 9. CATEGORY REVENUE CONCENTRATION
# ============================================

category_concentration = (
    category_performance
    .sort_values(
        "revenue",
        ascending=False
    )
    .reset_index(drop=True)
)

category_concentration["revenue_share_pct"] = (
    category_concentration["revenue"]
    /
    category_concentration["revenue"].sum()
    *
    100
)

category_concentration["cumulative_revenue_pct"] = (
    category_concentration["revenue_share_pct"]
    .cumsum()
)


print("\nCATEGORY REVENUE CONCENTRATION")

display(
    category_concentration[
        [
            "product_category_name",
            "revenue",
            "revenue_share_pct",
            "cumulative_revenue_pct"
        ]
    ]
)


# ============================================
# 10. TOP CATEGORY CONTRIBUTION
# ============================================

top_5_category_share = (
    category_concentration
    .head(5)["revenue_share_pct"]
    .sum()
)

top_10_category_share = (
    category_concentration
    .head(10)["revenue_share_pct"]
    .sum()
)


# ============================================
# 11. PRODUCT REVENUE CONCENTRATION
# ============================================

product_concentration = (
    product_summary
    .sort_values(
        "revenue",
        ascending=False
    )
    .reset_index(drop=True)
)

product_concentration["revenue_share_pct"] = (
    product_concentration["revenue"]
    /
    product_concentration["revenue"].sum()
    *
    100
)

product_concentration["cumulative_revenue_pct"] = (
    product_concentration["revenue_share_pct"]
    .cumsum()
)


top_10_product_share = (
    product_concentration
    .head(10)["revenue_share_pct"]
    .sum()
)


# ============================================
# 12. CATEGORY SUMMARY
# ============================================

print("\n" + "=" * 70)
print("PRODUCT & CATEGORY PERFORMANCE HIGHLIGHTS")
print("=" * 70)

print(
    f"Total Product Categories : "
    f"{category_performance['product_category_name'].nunique():,}"
)

print(
    f"Total Products Sold      : "
    f"{product_sales['product_id'].nunique():,}"
)

print(
    f"Top 5 Category Revenue % : "
    f"{top_5_category_share:.2f}%"
)

print(
    f"Top 10 Category Revenue %: "
    f"{top_10_category_share:.2f}%"
)

print(
    f"Top 10 Product Revenue % : "
    f"{top_10_product_share:.2f}%"
)

print(
    f"Top Category             : "
    f"{category_concentration.iloc[0]['product_category_name']}"
)

print(
    f"Top Category Revenue     : "
    f"{category_concentration.iloc[0]['revenue']:,.2f}"
)

print(
    f"Top Product              : "
    f"{product_concentration.iloc[0]['product_id']}"
)

print(
    f"Top Product Revenue      : "
    f"{product_concentration.iloc[0]['revenue']:,.2f}"
)


# ============================================
# 13. FINAL STATUS
# ============================================

print("\n" + "=" * 70)
print("SECTION 14.3 COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# SECTION 14.4 — CUSTOMER & CUSTOMER BEHAVIOR ANALYSIS
# ======================================================================

print("=" * 70)
print("SECTION 14.4 — CUSTOMER & CUSTOMER BEHAVIOR ANALYSIS")
print("=" * 70)


# ======================================================================
# 1. CUSTOMER OVERVIEW
# ======================================================================

print("\nCUSTOMER OVERVIEW")
print("-" * 70)

# Customer order counts
customer_orders = (
    dim_order
    .groupby("customer_unique_id")
    .agg(
        orders=("order_id", "nunique")
    )
    .reset_index()
)

# Customer revenue from order items
customer_revenue = (
    fact_order_items
    .merge(
        dim_order[["order_id", "customer_unique_id"]],
        on="order_id",
        how="left"
    )
    .groupby("customer_unique_id")
    .agg(
        order_items=("order_item_id", "count"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

# Combine customer metrics
customer_analysis = (
    customer_orders
    .merge(
        customer_revenue,
        on="customer_unique_id",
        how="left"
    )
)

customer_analysis["revenue"] = customer_analysis["revenue"].fillna(0)
customer_analysis["freight"] = customer_analysis["freight"].fillna(0)
customer_analysis["order_items"] = customer_analysis["order_items"].fillna(0)

customer_analysis["total_sales"] = (
    customer_analysis["revenue"] +
    customer_analysis["freight"]
)

customer_analysis["avg_order_value"] = (
    customer_analysis["revenue"] /
    customer_analysis["orders"]
)

customer_analysis["avg_items_per_order"] = (
    customer_analysis["order_items"] /
    customer_analysis["orders"]
)

# Customer type
customer_analysis["customer_type"] = customer_analysis["orders"].apply(
    lambda x: "Repeat Customer" if x > 1 else "One-Time Customer"
)

print(f"Total Customers        : {len(customer_analysis):,}")

print(
    f"One-Time Customers     : "
    f"{(customer_analysis['customer_type'] == 'One-Time Customer').sum():,}"
)

print(
    f"Repeat Customers       : "
    f"{(customer_analysis['customer_type'] == 'Repeat Customer').sum():,}"
)

print(
    f"Average Orders/Customer: "
    f"{customer_analysis['orders'].mean():.2f}"
)

print(
    f"Average Revenue/Customer: "
    f"{customer_analysis['revenue'].mean():.2f}"
)


# ======================================================================
# 2. CUSTOMER ORDER FREQUENCY
# ======================================================================

print("\nCUSTOMER ORDER FREQUENCY")
print("-" * 70)

order_frequency = (
    customer_analysis
    .groupby("orders")
    .agg(
        customers=("customer_unique_id", "nunique")
    )
    .reset_index()
    .sort_values("orders")
)

order_frequency["customer_share_pct"] = (
    order_frequency["customers"] /
    len(customer_analysis) * 100
)

print(order_frequency.to_string(index=False))


# ======================================================================
# 3. TOP 20 CUSTOMERS BY REVENUE
# ======================================================================

print("\nTOP 20 CUSTOMERS BY REVENUE")
print("-" * 70)

top_customers_revenue = (
    customer_analysis
    .sort_values("revenue", ascending=False)
    .head(20)
    .copy()
)

top_customers_revenue = top_customers_revenue[
    [
        "customer_unique_id",
        "orders",
        "order_items",
        "revenue",
        "freight",
        "total_sales",
        "avg_order_value",
        "customer_type"
    ]
]

print(top_customers_revenue.to_string(index=False))


# ======================================================================
# 4. TOP 20 CUSTOMERS BY ORDER COUNT
# ======================================================================

print("\nTOP 20 CUSTOMERS BY ORDER COUNT")
print("-" * 70)

top_customers_orders = (
    customer_analysis
    .sort_values(
        ["orders", "revenue"],
        ascending=[False, False]
    )
    .head(20)
    .copy()
)

top_customers_orders = top_customers_orders[
    [
        "customer_unique_id",
        "orders",
        "order_items",
        "revenue",
        "freight",
        "total_sales",
        "avg_order_value",
        "customer_type"
    ]
]

print(top_customers_orders.to_string(index=False))


# ======================================================================
# 5. CUSTOMER REVENUE CONCENTRATION
# ======================================================================

print("\nCUSTOMER REVENUE CONCENTRATION")
print("-" * 70)

customer_concentration = (
    customer_analysis
    .sort_values("revenue", ascending=False)
    [
        [
            "customer_unique_id",
            "revenue"
        ]
    ]
    .copy()
)

total_customer_revenue = customer_concentration["revenue"].sum()

customer_concentration["revenue_share_pct"] = (
    customer_concentration["revenue"] /
    total_customer_revenue * 100
)

customer_concentration["cumulative_revenue_pct"] = (
    customer_concentration["revenue_share_pct"].cumsum()
)

print(
    customer_concentration
    .head(20)
    .to_string(index=False)
)


# ======================================================================
# 6. CUSTOMER SEGMENT ANALYSIS
# ======================================================================

print("\nCUSTOMER SEGMENT ANALYSIS")
print("-" * 70)

customer_segments = (
    customer_analysis
    .groupby("customer_type")
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("orders", "sum"),
        order_items=("order_items", "sum"),
        revenue=("revenue", "sum"),
        freight=("freight", "sum"),
        total_sales=("total_sales", "sum")
    )
    .reset_index()
)

customer_segments["customer_share_pct"] = (
    customer_segments["customers"] /
    customer_segments["customers"].sum() * 100
)

customer_segments["revenue_share_pct"] = (
    customer_segments["revenue"] /
    customer_segments["revenue"].sum() * 100
)

customer_segments["avg_revenue_per_customer"] = (
    customer_segments["revenue"] /
    customer_segments["customers"]
)

customer_segments["avg_orders_per_customer"] = (
    customer_segments["orders"] /
    customer_segments["customers"]
)

print(customer_segments.to_string(index=False))


# ======================================================================
# 7. HIGH-VALUE CUSTOMER ANALYSIS
# ======================================================================

print("\nHIGH-VALUE CUSTOMER ANALYSIS")
print("-" * 70)

# Top 10% customers by revenue
revenue_threshold = customer_analysis["revenue"].quantile(0.90)

high_value_customers = customer_analysis[
    customer_analysis["revenue"] >= revenue_threshold
].copy()

high_value_revenue_share = (
    high_value_customers["revenue"].sum() /
    customer_analysis["revenue"].sum() * 100
)

print(
    f"Top 10% Revenue Threshold : "
    f"{revenue_threshold:.2f}"
)

print(
    f"High-Value Customers      : "
    f"{len(high_value_customers):,}"
)

print(
    f"Revenue Contribution      : "
    f"{high_value_revenue_share:.2f}%"
)


# ======================================================================
# 8. REPEAT CUSTOMER ANALYSIS
# ======================================================================

print("\nREPEAT CUSTOMER ANALYSIS")
print("-" * 70)

repeat_customers = customer_analysis[
    customer_analysis["orders"] > 1
].copy()

one_time_customers = customer_analysis[
    customer_analysis["orders"] == 1
].copy()

repeat_customer_rate = (
    len(repeat_customers) /
    len(customer_analysis) * 100
)

repeat_revenue_share = (
    repeat_customers["revenue"].sum() /
    customer_analysis["revenue"].sum() * 100
)

print(
    f"Repeat Customers          : "
    f"{len(repeat_customers):,}"
)

print(
    f"One-Time Customers        : "
    f"{len(one_time_customers):,}"
)

print(
    f"Repeat Customer Rate      : "
    f"{repeat_customer_rate:.2f}%"
)

print(
    f"Repeat Customer Revenue % : "
    f"{repeat_revenue_share:.2f}%"
)


# ======================================================================
# 9. CUSTOMER GEOGRAPHICAL ANALYSIS
# ======================================================================

print("\nCUSTOMER GEOGRAPHICAL ANALYSIS")
print("-" * 70)

customer_geo = (
    customer_analysis
    .merge(
        dim_customer[
            [
                "customer_unique_id",
                "customer_city",
                "customer_state"
            ]
        ],
        on="customer_unique_id",
        how="left"
    )
)

customer_state_analysis = (
    customer_geo
    .groupby("customer_state")
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("orders", "sum"),
        revenue=("revenue", "sum"),
        freight=("freight", "sum"),
        total_sales=("total_sales", "sum")
    )
    .reset_index()
)

customer_state_analysis["revenue_share_pct"] = (
    customer_state_analysis["revenue"] /
    customer_state_analysis["revenue"].sum() * 100
)

customer_state_analysis["avg_revenue_per_customer"] = (
    customer_state_analysis["revenue"] /
    customer_state_analysis["customers"]
)

customer_state_analysis = (
    customer_state_analysis
    .sort_values("revenue", ascending=False)
)

print(
    customer_state_analysis
    .head(20)
    .to_string(index=False)
)


# ======================================================================
# 10. TOP STATES BY CUSTOMER REVENUE
# ======================================================================

print("\nTOP STATES BY CUSTOMER REVENUE")
print("-" * 70)

print(
    customer_state_analysis
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# 11. TOP STATES BY CUSTOMER COUNT
# ======================================================================

print("\nTOP STATES BY CUSTOMER COUNT")
print("-" * 70)

top_states_customers = (
    customer_state_analysis
    .sort_values("customers", ascending=False)
    .head(10)
)

print(
    top_states_customers
    .to_string(index=False)
)


# ======================================================================
# 12. CUSTOMER BEHAVIOR HIGHLIGHTS
# ======================================================================

print("\n" + "=" * 70)
print("CUSTOMER BEHAVIOR HIGHLIGHTS")
print("=" * 70)

highest_value_customer = (
    customer_analysis
    .sort_values("revenue", ascending=False)
    .iloc[0]
)

most_frequent_customer = (
    customer_analysis
    .sort_values(
        ["orders", "revenue"],
        ascending=[False, False]
    )
    .iloc[0]
)

print(
    f"Total Customers           : "
    f"{len(customer_analysis):,}"
)

print(
    f"Repeat Customers          : "
    f"{len(repeat_customers):,}"
)

print(
    f"Repeat Customer Rate      : "
    f"{repeat_customer_rate:.2f}%"
)

print(
    f"Highest-Value Customer    : "
    f"{highest_value_customer['customer_unique_id']}"
)

print(
    f"Highest Customer Revenue  : "
    f"{highest_value_customer['revenue']:,.2f}"
)

print(
    f"Most Frequent Customer    : "
    f"{most_frequent_customer['customer_unique_id']}"
)

print(
    f"Most Customer Orders      : "
    f"{int(most_frequent_customer['orders']):,}"
)

print(
    f"Average Revenue/Customer  : "
    f"{customer_analysis['revenue'].mean():,.2f}"
)

print(
    f"Average Orders/Customer   : "
    f"{customer_analysis['orders'].mean():.2f}"
)

print(
    f"Top 10% Revenue Share     : "
    f"{high_value_revenue_share:.2f}%"
)

top_state = customer_state_analysis.iloc[0]

print(
    f"Top Revenue State         : "
    f"{top_state['customer_state']}"
)

print(
    f"Top State Revenue         : "
    f"{top_state['revenue']:,.2f}"
)


# ======================================================================
# 13. FINAL SECTION STATUS
# ======================================================================

print("\n" + "=" * 70)
print("SECTION 14.4 COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# SECTION 14.5 — CUSTOMER RETENTION & RFM ANALYSIS
# ======================================================================

print("=" * 70)
print("SECTION 14.5 — CUSTOMER RETENTION & RFM ANALYSIS")
print("=" * 70)


# ======================================================================
# 1. CUSTOMER PURCHASE HISTORY
# ======================================================================

print("\nCUSTOMER PURCHASE HISTORY")
print("-" * 70)

customer_purchase_history = (
    dim_order
    .groupby("customer_unique_id")
    .agg(
        first_purchase=("order_purchase_date", "min"),
        last_purchase=("order_purchase_date", "max"),
        orders=("order_id", "nunique")
    )
    .reset_index()
)

print(f"Total Customers : {len(customer_purchase_history):,}")
print(customer_purchase_history.head().to_string(index=False))


# ======================================================================
# 2. CUSTOMER RECENCY
# ======================================================================

print("\nCUSTOMER RECENCY")
print("-" * 70)

analysis_date = dim_order["order_purchase_date"].max()

customer_purchase_history["recency_days"] = (
    analysis_date -
    customer_purchase_history["last_purchase"]
).dt.days

print(
    customer_purchase_history["recency_days"]
    .describe()
    .to_string()
)


# ======================================================================
# 3. CUSTOMER FREQUENCY
# ======================================================================

print("\nCUSTOMER FREQUENCY")
print("-" * 70)

customer_purchase_history["frequency"] = (
    customer_purchase_history["orders"]
)

print(
    customer_purchase_history["frequency"]
    .describe()
    .to_string()
)


# ======================================================================
# 4. CUSTOMER MONETARY VALUE
# ======================================================================

print("\nCUSTOMER MONETARY VALUE")
print("-" * 70)

customer_monetary = (
    fact_order_items
    .merge(
        dim_order[
            [
                "order_id",
                "customer_unique_id"
            ]
        ],
        on="order_id",
        how="left"
    )
    .groupby("customer_unique_id")
    .agg(
        monetary_value=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

customer_monetary["total_value"] = (
    customer_monetary["monetary_value"] +
    customer_monetary["freight"]
)

customer_rfm = (
    customer_purchase_history
    .merge(
        customer_monetary,
        on="customer_unique_id",
        how="left"
    )
)

customer_rfm["monetary_value"] = (
    customer_rfm["monetary_value"].fillna(0)
)

customer_rfm["freight"] = (
    customer_rfm["freight"].fillna(0)
)

customer_rfm["total_value"] = (
    customer_rfm["total_value"].fillna(0)
)


# ======================================================================
# 5. RFM SCORE
# ======================================================================

print("\nRFM SCORE")
print("-" * 70)

# Recency:
# Lower recency = better customer
customer_rfm["R_score"] = pd.qcut(
    customer_rfm["recency_days"],
    5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
)

# Frequency:
# Higher frequency = better customer
customer_rfm["F_score"] = pd.qcut(
    customer_rfm["frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

# Monetary:
# Higher monetary value = better customer
customer_rfm["M_score"] = pd.qcut(
    customer_rfm["monetary_value"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

customer_rfm["R_score"] = customer_rfm["R_score"].astype(int)
customer_rfm["F_score"] = customer_rfm["F_score"].astype(int)
customer_rfm["M_score"] = customer_rfm["M_score"].astype(int)

customer_rfm["RFM_score"] = (
    customer_rfm["R_score"].astype(str) +
    customer_rfm["F_score"].astype(str) +
    customer_rfm["M_score"].astype(str)
)

customer_rfm["RFM_total"] = (
    customer_rfm["R_score"] +
    customer_rfm["F_score"] +
    customer_rfm["M_score"]
)


# ======================================================================
# 6. CUSTOMER SEGMENTS
# ======================================================================

print("\nCUSTOMER SEGMENTS")
print("-" * 70)

def classify_customer(row):

    r = row["R_score"]
    f = row["F_score"]
    m = row["M_score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    elif r >= 4 and f >= 3:
        return "Loyal Customers"

    elif r >= 4 and m >= 4:
        return "High Value New Customers"

    elif r >= 3 and f >= 3:
        return "Potential Loyalists"

    elif r <= 2 and f >= 4:
        return "At Risk Loyal Customers"

    elif r <= 2 and m >= 4:
        return "High Value At Risk"

    elif r <= 2:
        return "Lost Customers"

    else:
        return "Needs Attention"


customer_rfm["customer_segment"] = (
    customer_rfm.apply(
        classify_customer,
        axis=1
    )
)

rfm_segments = (
    customer_rfm
    .groupby("customer_segment")
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("orders", "sum"),
        revenue=("monetary_value", "sum"),
        total_value=("total_value", "sum"),
        avg_recency=("recency_days", "mean"),
        avg_frequency=("frequency", "mean"),
        avg_customer_value=("monetary_value", "mean")
    )
    .reset_index()
)

rfm_segments["customer_share_pct"] = (
    rfm_segments["customers"] /
    rfm_segments["customers"].sum() * 100
)

rfm_segments["revenue_share_pct"] = (
    rfm_segments["revenue"] /
    rfm_segments["revenue"].sum() * 100
)

rfm_segments = rfm_segments.sort_values(
    "revenue",
    ascending=False
)

print(rfm_segments.to_string(index=False))


# ======================================================================
# 7. CHAMPION CUSTOMERS
# ======================================================================

print("\nCHAMPION CUSTOMERS")
print("-" * 70)

champion_customers = (
    customer_rfm[
        customer_rfm["customer_segment"] == "Champions"
    ]
    .sort_values(
        ["monetary_value", "frequency"],
        ascending=[False, False]
    )
    .head(20)
)

print(
    champion_customers[
        [
            "customer_unique_id",
            "orders",
            "recency_days",
            "frequency",
            "monetary_value",
            "RFM_score"
        ]
    ]
    .to_string(index=False)
)


# ======================================================================
# 8. AT-RISK CUSTOMERS
# ======================================================================

print("\nAT-RISK CUSTOMERS")
print("-" * 70)

at_risk_customers = customer_rfm[
    customer_rfm["customer_segment"].isin(
        [
            "At Risk Loyal Customers",
            "High Value At Risk"
        ]
    )
].copy()

print(
    f"At-Risk Customers : "
    f"{len(at_risk_customers):,}"
)

print(
    at_risk_customers[
        [
            "customer_unique_id",
            "orders",
            "recency_days",
            "frequency",
            "monetary_value",
            "customer_segment"
        ]
    ]
    .sort_values(
        "monetary_value",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)


# ======================================================================
# 9. CUSTOMER RETENTION GROUPS
# ======================================================================

print("\nCUSTOMER RETENTION GROUPS")
print("-" * 70)

customer_rfm["retention_group"] = pd.cut(
    customer_rfm["recency_days"],
    bins=[
        -1,
        30,
        90,
        180,
        365,
        float("inf")
    ],
    labels=[
        "Active - 0 to 30 Days",
        "Recent - 31 to 90 Days",
        "Dormant - 91 to 180 Days",
        "Inactive - 181 to 365 Days",
        "Lost - 365+ Days"
    ]
)

retention_analysis = (
    customer_rfm
    .groupby(
        "retention_group",
        observed=False
    )
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("orders", "sum"),
        revenue=("monetary_value", "sum"),
        avg_customer_value=("monetary_value", "mean")
    )
    .reset_index()
)

retention_analysis["customer_share_pct"] = (
    retention_analysis["customers"] /
    retention_analysis["customers"].sum() * 100
)

retention_analysis["revenue_share_pct"] = (
    retention_analysis["revenue"] /
    retention_analysis["revenue"].sum() * 100
)

print(
    retention_analysis.to_string(index=False)
)


# ======================================================================
# 10. RFM SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("RFM SUMMARY")
print("=" * 70)

print(
    f"Total Customers       : "
    f"{len(customer_rfm):,}"
)

print(
    f"Average Recency       : "
    f"{customer_rfm['recency_days'].mean():.2f} days"
)

print(
    f"Average Frequency     : "
    f"{customer_rfm['frequency'].mean():.2f} orders"
)

print(
    f"Average Monetary Value: "
    f"{customer_rfm['monetary_value'].mean():,.2f}"
)

print(
    f"Champions             : "
    f"{(customer_rfm['customer_segment'] == 'Champions').sum():,}"
)

print(
    f"At-Risk Customers     : "
    f"{len(at_risk_customers):,}"
)

print(
    f"High Value At Risk    : "
    f"{(customer_rfm['customer_segment'] == 'High Value At Risk').sum():,}"
)

print(
    f"Lost Customers        : "
    f"{(customer_rfm['customer_segment'] == 'Lost Customers').sum():,}"
)


# ======================================================================
# 11. FINAL SECTION STATUS
# ======================================================================

print("\n" + "=" * 70)
print("SECTION 14.5 COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# SECTION 14.6 — DELIVERY & LOGISTICS ANALYSIS
# ======================================================================

print("=" * 70)
print("SECTION 14.6 — DELIVERY & LOGISTICS ANALYSIS")
print("=" * 70)


# ======================================================================
# 1. DELIVERY OVERVIEW
# ======================================================================

print("\nDELIVERY OVERVIEW")
print("-" * 70)

delivery_data = dim_order.copy()

delivery_data["order_purchase_date"] = pd.to_datetime(
    delivery_data["order_purchase_date"]
)

delivery_data["order_delivered_customer_date"] = pd.to_datetime(
    delivery_data["order_delivered_customer_date"]
)

delivery_data["order_estimated_delivery_date"] = pd.to_datetime(
    delivery_data["order_estimated_delivery_date"]
)

# Actual delivery time
delivery_data["delivery_days"] = (
    delivery_data["order_delivered_customer_date"]
    - delivery_data["order_purchase_date"]
).dt.total_seconds() / 86400

# Estimated delivery time
delivery_data["estimated_delivery_days"] = (
    delivery_data["order_estimated_delivery_date"]
    - delivery_data["order_purchase_date"]
).dt.total_seconds() / 86400

# Difference between actual and estimated
delivery_data["delivery_difference_days"] = (
    delivery_data["order_delivered_customer_date"]
    - delivery_data["order_estimated_delivery_date"]
).dt.total_seconds() / 86400


print(
    f"Total Orders                  : "
    f"{len(delivery_data):,}"
)

print(
    f"Orders With Delivery Date     : "
    f"{delivery_data['order_delivered_customer_date'].notna().sum():,}"
)

print(
    f"Average Delivery Time         : "
    f"{delivery_data['delivery_days'].mean():.2f} days"
)

print(
    f"Median Delivery Time          : "
    f"{delivery_data['delivery_days'].median():.2f} days"
)

print(
    f"Average Estimated Delivery    : "
    f"{delivery_data['estimated_delivery_days'].mean():.2f} days"
)


# ======================================================================
# 2. DELIVERY STATUS
# ======================================================================

print("\nDELIVERY STATUS")
print("-" * 70)

delivery_data["delivery_status"] = np.select(
    [
        delivery_data["order_delivered_customer_date"].isna(),
        delivery_data["delivery_difference_days"] > 0,
        delivery_data["delivery_difference_days"] <= 0
    ],
    [
        "Not Delivered",
        "Late",
        "On Time / Early"
    ],
    default="Unknown"
)

delivery_status = (
    delivery_data
    .groupby("delivery_status")
    .agg(
        orders=("order_id", "nunique")
    )
    .reset_index()
)

delivery_status["order_share_pct"] = (
    delivery_status["orders"] /
    delivery_status["orders"].sum() * 100
)

print(
    delivery_status
    .sort_values("orders", ascending=False)
    .to_string(index=False)
)


# ======================================================================
# 3. ON-TIME DELIVERY PERFORMANCE
# ======================================================================

print("\nON-TIME DELIVERY PERFORMANCE")
print("-" * 70)

delivered_orders = delivery_data[
    delivery_data["order_delivered_customer_date"].notna()
].copy()

total_delivered = len(delivered_orders)

on_time_orders = (
    delivered_orders["delivery_difference_days"] <= 0
).sum()

late_orders = (
    delivered_orders["delivery_difference_days"] > 0
).sum()

on_time_rate = (
    on_time_orders / total_delivered * 100
    if total_delivered > 0 else 0
)

late_rate = (
    late_orders / total_delivered * 100
    if total_delivered > 0 else 0
)

print(f"Delivered Orders : {total_delivered:,}")
print(f"On-Time / Early  : {on_time_orders:,}")
print(f"Late Orders      : {late_orders:,}")
print(f"On-Time Rate     : {on_time_rate:.2f}%")
print(f"Late Rate        : {late_rate:.2f}%")


# ======================================================================
# 4. DELIVERY TIME DISTRIBUTION
# ======================================================================

print("\nDELIVERY TIME DISTRIBUTION")
print("-" * 70)

delivery_data["delivery_bucket"] = pd.cut(
    delivery_data["delivery_days"],
    bins=[
        -1,
        3,
        7,
        14,
        21,
        30,
        60,
        float("inf")
    ],
    labels=[
        "0-3 Days",
        "4-7 Days",
        "8-14 Days",
        "15-21 Days",
        "22-30 Days",
        "31-60 Days",
        "60+ Days"
    ]
)

delivery_distribution = (
    delivery_data
    .groupby(
        "delivery_bucket",
        observed=False
    )
    .agg(
        orders=("order_id", "nunique")
    )
    .reset_index()
)

delivery_distribution["order_share_pct"] = (
    delivery_distribution["orders"] /
    delivery_distribution["orders"].sum() * 100
)

print(
    delivery_distribution.to_string(index=False)
)


# ======================================================================
# 5. LATE DELIVERY ANALYSIS
# ======================================================================

print("\nLATE DELIVERY ANALYSIS")
print("-" * 70)

late_delivery = delivered_orders[
    delivered_orders["delivery_difference_days"] > 0
].copy()

print(
    f"Late Orders             : "
    f"{len(late_delivery):,}"
)

print(
    f"Average Delay           : "
    f"{late_delivery['delivery_difference_days'].mean():.2f} days"
)

print(
    f"Maximum Delay           : "
    f"{late_delivery['delivery_difference_days'].max():.2f} days"
)

print(
    f"Median Delay            : "
    f"{late_delivery['delivery_difference_days'].median():.2f} days"
)


# ======================================================================
# 6. DELIVERY PERFORMANCE BY CUSTOMER STATE
# ======================================================================

print("\nDELIVERY PERFORMANCE BY CUSTOMER STATE")
print("-" * 70)

# Get customer state
state_mapping = dim_customer[
    [
        "customer_unique_id",
        "customer_state"
    ]
].drop_duplicates()

delivery_state = (
    delivery_data[
        [
            "order_id",
            "customer_unique_id",
            "delivery_days",
            "delivery_difference_days",
            "delivery_status"
        ]
    ]
    .merge(
        state_mapping,
        on="customer_unique_id",
        how="left"
    )
)

state_delivery_analysis = (
    delivery_state
    .groupby("customer_state")
    .agg(
        orders=("order_id", "nunique"),
        avg_delivery_days=("delivery_days", "mean"),
        median_delivery_days=("delivery_days", "median"),
        avg_delay_days=("delivery_difference_days", "mean"),
        late_orders=(
            "delivery_status",
            lambda x: (x == "Late").sum()
        ),
        on_time_orders=(
            "delivery_status",
            lambda x: (x == "On Time / Early").sum()
        )
    )
    .reset_index()
)

state_delivery_analysis["late_rate_pct"] = (
    state_delivery_analysis["late_orders"] /
    (
        state_delivery_analysis["late_orders"] +
        state_delivery_analysis["on_time_orders"]
    ) * 100
)

state_delivery_analysis["on_time_rate_pct"] = (
    state_delivery_analysis["on_time_orders"] /
    (
        state_delivery_analysis["late_orders"] +
        state_delivery_analysis["on_time_orders"]
    ) * 100
)

print(
    state_delivery_analysis
    .sort_values(
        "orders",
        ascending=False
    )
    .to_string(index=False)
)


# ======================================================================
# 7. WORST STATES BY DELIVERY TIME
# ======================================================================

print("\nWORST STATES BY AVERAGE DELIVERY TIME")
print("-" * 70)

print(
    state_delivery_analysis
    .sort_values(
        "avg_delivery_days",
        ascending=False
    )
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# 8. WORST STATES BY LATE DELIVERY RATE
# ======================================================================

print("\nWORST STATES BY LATE DELIVERY RATE")
print("-" * 70)

print(
    state_delivery_analysis
    .sort_values(
        "late_rate_pct",
        ascending=False
    )
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# 9. FREIGHT COST ANALYSIS
# ======================================================================

print("\nFREIGHT COST ANALYSIS")
print("-" * 70)

# Aggregate order-level freight
freight_analysis = (
    fact_order_items
    .groupby("order_id")
    .agg(
        order_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
    .reset_index()
)

freight_analysis["freight_pct"] = np.where(
    freight_analysis["order_value"] > 0,
    freight_analysis["freight_value"] /
    freight_analysis["order_value"] * 100,
    np.nan
)

print(
    f"Total Product Value : "
    f"{freight_analysis['order_value'].sum():,.2f}"
)

print(
    f"Total Freight       : "
    f"{freight_analysis['freight_value'].sum():,.2f}"
)

print(
    f"Average Freight/Order : "
    f"{freight_analysis['freight_value'].mean():,.2f}"
)

print(
    f"Average Freight %     : "
    f"{freight_analysis['freight_pct'].mean():.2f}%"
)

print(
    f"Median Freight %      : "
    f"{freight_analysis['freight_pct'].median():.2f}%"
)


# ======================================================================
# 10. FREIGHT COST BY STATE
# ======================================================================

print("\nFREIGHT COST BY CUSTOMER STATE")
print("-" * 70)

freight_state = (
    freight_analysis
    .merge(
        dim_order[
            [
                "order_id",
                "customer_unique_id"
            ]
        ],
        on="order_id",
        how="left"
    )
    .merge(
        state_mapping,
        on="customer_unique_id",
        how="left"
    )
)

state_freight_analysis = (
    freight_state
    .groupby("customer_state")
    .agg(
        orders=("order_id", "nunique"),
        order_value=("order_value", "sum"),
        freight=("freight_value", "sum"),
        avg_freight=("freight_value", "mean")
    )
    .reset_index()
)

state_freight_analysis["freight_pct"] = (
    state_freight_analysis["freight"] /
    state_freight_analysis["order_value"] * 100
)

print(
    state_freight_analysis
    .sort_values(
        "freight",
        ascending=False
    )
    .to_string(index=False)
)


# ======================================================================
# 11. TOP HIGH-FREIGHT STATES
# ======================================================================

print("\nTOP STATES BY FREIGHT COST")
print("-" * 70)

print(
    state_freight_analysis
    .sort_values(
        "freight",
        ascending=False
    )
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# 12. DELIVERY + FREIGHT PERFORMANCE
# ======================================================================

print("\nDELIVERY + FREIGHT PERFORMANCE")
print("-" * 70)

delivery_freight = (
    state_delivery_analysis
    .merge(
        state_freight_analysis,
        on="customer_state",
        how="left"
    )
)

print(
    delivery_freight[
        [
            "customer_state",
            "orders_x",
            "avg_delivery_days",
            "late_rate_pct",
            "on_time_rate_pct",
            "freight",
            "freight_pct"
        ]
    ]
    .sort_values(
        "late_rate_pct",
        ascending=False
    )
    .to_string(index=False)
)


# ======================================================================
# 13. EXTREME DELIVERY CASES
# ======================================================================

print("\nEXTREME DELIVERY CASES")
print("-" * 70)

print("\nLongest Delivery Times:")

print(
    delivered_orders[
        [
            "order_id",
            "customer_unique_id",
            "delivery_days",
            "delivery_difference_days"
        ]
    ]
    .sort_values(
        "delivery_days",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)


print("\nLargest Delivery Delays:")

print(
    late_delivery[
        [
            "order_id",
            "customer_unique_id",
            "delivery_days",
            "delivery_difference_days"
        ]
    ]
    .sort_values(
        "delivery_difference_days",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)


# ======================================================================
# 14. DELIVERY SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("DELIVERY & LOGISTICS SUMMARY")
print("=" * 70)

print(
    f"Total Orders              : "
    f"{len(delivery_data):,}"
)

print(
    f"Delivered Orders          : "
    f"{total_delivered:,}"
)

print(
    f"Average Delivery Time     : "
    f"{delivery_data['delivery_days'].mean():.2f} days"
)

print(
    f"On-Time Delivery Rate     : "
    f"{on_time_rate:.2f}%"
)

print(
    f"Late Delivery Rate        : "
    f"{late_rate:.2f}%"
)

print(
    f"Average Delay             : "
    f"{late_delivery['delivery_difference_days'].mean():.2f} days"
)

print(
    f"Total Freight             : "
    f"{freight_analysis['freight_value'].sum():,.2f}"
)

print(
    f"Average Freight/Order     : "
    f"{freight_analysis['freight_value'].mean():,.2f}"
)

print(
    f"Average Freight %         : "
    f"{freight_analysis['freight_pct'].mean():.2f}%"
)


# ======================================================================
# 15. FINAL SECTION STATUS
# ======================================================================

print("\n" + "=" * 70)
print("SECTION 14.6 COMPLETED")
print("=" * 70)
```

```python
# ==============================================================
# SECTION 14.7 — STEP 2: REVIEW SCORE DISTRIBUTION
# ==============================================================

review_distribution = (
    order_reviews["review_score"]
    .value_counts()
    .sort_index()
    .reset_index()
)

review_distribution.columns = ["review_score", "reviews"]

review_distribution["review_share_pct"] = (
    review_distribution["reviews"] / len(order_reviews) * 100
)

print("\nREVIEW SCORE DISTRIBUTION")
print("-" * 70)
print(review_distribution.to_string(index=False))

# Satisfaction groups
review_satisfaction = pd.DataFrame({
    "satisfaction_level": [
        "Very Dissatisfied",
        "Dissatisfied",
        "Neutral",
        "Satisfied",
        "Very Satisfied"
    ],
    "reviews": [
        (order_reviews["review_score"] == 1).sum(),
        (order_reviews["review_score"] == 2).sum(),
        (order_reviews["review_score"] == 3).sum(),
        (order_reviews["review_score"] == 4).sum(),
        (order_reviews["review_score"] == 5).sum()
    ]
})

review_satisfaction["share_pct"] = (
    review_satisfaction["reviews"] / len(order_reviews) * 100
)

print("\nCUSTOMER SATISFACTION LEVEL")
print("-" * 70)
print(review_satisfaction.to_string(index=False))
```

```python
# ==============================================================
# SECTION 14.7 — STEP 3: CUSTOMER SATISFACTION KPIs
# ==============================================================

total_reviews = len(order_reviews)

five_star_reviews = (order_reviews["review_score"] == 5).sum()
four_five_reviews = order_reviews["review_score"].isin([4, 5]).sum()
one_two_reviews = order_reviews["review_score"].isin([1, 2]).sum()
one_star_reviews = (order_reviews["review_score"] == 1).sum()

satisfaction_kpis = pd.DataFrame({
    "metric": [
        "Average Review Score",
        "5-Star Rate",
        "4-5 Star Satisfaction Rate",
        "1-2 Star Negative Review Rate",
        "1-Star Rate",
        "3-Star Neutral Rate"
    ],
    "value": [
        order_reviews["review_score"].mean(),
        five_star_reviews / total_reviews * 100,
        four_five_reviews / total_reviews * 100,
        one_two_reviews / total_reviews * 100,
        one_star_reviews / total_reviews * 100,
        (order_reviews["review_score"] == 3).sum() / total_reviews * 100
    ]
})

print("\nCUSTOMER SATISFACTION KPIs")
print("-" * 70)
print(satisfaction_kpis.to_string(index=False))
```

```python
# ==============================================================
# SECTION 14.7 — STEP 4: REVIEW SCORE BY CUSTOMER STATE
# ==============================================================

# Merge review data with order/customer information
review_state = order_reviews.merge(
    orders[["order_id", "customer_id"]],
    on="order_id",
    how="left"
).merge(
    customers[["customer_id", "customer_state"]],
    on="customer_id",
    how="left"
)

review_state_analysis = (
    review_state
    .groupby("customer_state")
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        five_star_reviews=("review_score", lambda x: (x == 5).sum()),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum()),
        one_star_reviews=("review_score", lambda x: (x == 1).sum())
    )
    .reset_index()
)

review_state_analysis["five_star_rate_pct"] = (
    review_state_analysis["five_star_reviews"]
    / review_state_analysis["reviews"] * 100
)

review_state_analysis["negative_review_rate_pct"] = (
    review_state_analysis["negative_reviews"]
    / review_state_analysis["reviews"] * 100
)

review_state_analysis["one_star_rate_pct"] = (
    review_state_analysis["one_star_reviews"]
    / review_state_analysis["reviews"] * 100
)

review_state_analysis = review_state_analysis.sort_values(
    "reviews", ascending=False
)

print("\nREVIEW PERFORMANCE BY CUSTOMER STATE")
print("-" * 70)
print(review_state_analysis.to_string(index=False))

print("\nLOWEST RATED STATES")
print("-" * 70)
print(
    review_state_analysis
    .sort_values("avg_review_score")
    .head(10)
    .to_string(index=False)
)

print("\nHIGHEST RATED STATES")
print("-" * 70)
print(
    review_state_analysis
    .sort_values("avg_review_score", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
# ==============================================================
# SECTION 14.7 — STEP 5: REVIEW SCORE VS DELIVERY PERFORMANCE
# ==============================================================

# Prepare delivery data
delivery_review = orders[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].copy()

# Convert dates
delivery_review["order_delivered_customer_date"] = pd.to_datetime(
    delivery_review["order_delivered_customer_date"],
    errors="coerce"
)

delivery_review["order_estimated_delivery_date"] = pd.to_datetime(
    delivery_review["order_estimated_delivery_date"],
    errors="coerce"
)

# Calculate delivery difference
delivery_review["delivery_difference_days"] = (
    delivery_review["order_delivered_customer_date"]
    - delivery_review["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Create delivery status
delivery_review["delivery_status"] = np.select(
    [
        delivery_review["order_delivered_customer_date"].isna(),
        delivery_review["delivery_difference_days"] <= 0,
        delivery_review["delivery_difference_days"] > 0
    ],
    [
        "Not Delivered",
        "On Time / Early",
        "Late"
    ],
    default="Unknown"
)

# Merge with reviews
review_delivery = order_reviews[
    ["order_id", "review_score"]
].merge(
    delivery_review[
        ["order_id", "delivery_status", "delivery_difference_days"]
    ],
    on="order_id",
    how="left"
)

# Average score by delivery status
delivery_review_analysis = (
    review_delivery
    .groupby("delivery_status")
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        five_star_reviews=("review_score", lambda x: (x == 5).sum()),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum())
    )
    .reset_index()
)

delivery_review_analysis["five_star_rate_pct"] = (
    delivery_review_analysis["five_star_reviews"]
    / delivery_review_analysis["reviews"] * 100
)

delivery_review_analysis["negative_review_rate_pct"] = (
    delivery_review_analysis["negative_reviews"]
    / delivery_review_analysis["reviews"] * 100
)

print("\nREVIEW SCORE VS DELIVERY PERFORMANCE")
print("-" * 70)
print(
    delivery_review_analysis
    .to_string(index=False)
)

# Average review score by delivery delay bucket
review_delivery["delay_bucket"] = pd.cut(
    review_delivery["delivery_difference_days"],
    bins=[-np.inf, -15, -7, 0, 3, 7, 14, 30, np.inf],
    labels=[
        "15+ Days Early",
        "8-14 Days Early",
        "1-7 Days Early",
        "1-3 Days Late",
        "4-7 Days Late",
        "8-14 Days Late",
        "15-30 Days Late",
        "30+ Days Late"
    ]
)

delay_review_analysis = (
    review_delivery
    .dropna(subset=["delay_bucket"])
    .groupby("delay_bucket", observed=False)
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum())
    )
    .reset_index()
)

delay_review_analysis["negative_review_rate_pct"] = (
    delay_review_analysis["negative_reviews"]
    / delay_review_analysis["reviews"] * 100
)

print("\nREVIEW SCORE BY DELIVERY DELAY")
print("-" * 70)
print(
    delay_review_analysis
    .to_string(index=False)
)
```

```python
# ==============================================================
# SECTION 14.7 — STEP 6: REVIEW SCORE VS FREIGHT COST
# ==============================================================

# Get freight/order value information
freight_review = order_items[
    ["order_id", "price", "freight_value"]
].copy()

# Aggregate item-level data to order level
freight_order = (
    freight_review
    .groupby("order_id")
    .agg(
        product_value=("price", "sum"),
        freight=("freight_value", "sum")
    )
    .reset_index()
)

# Calculate freight percentage
freight_order["freight_pct"] = np.where(
    freight_order["product_value"] > 0,
    freight_order["freight"] / freight_order["product_value"] * 100,
    np.nan
)

# Merge with reviews
review_freight = order_reviews[
    ["order_id", "review_score"]
].merge(
    freight_order,
    on="order_id",
    how="left"
)

# --------------------------------------------------------------
# Review score by freight cost bucket
# --------------------------------------------------------------

review_freight["freight_bucket"] = pd.cut(
    review_freight["freight"],
    bins=[-np.inf, 10, 20, 30, 50, 100, np.inf],
    labels=[
        "≤ ₹10",
        "₹10-20",
        "₹20-30",
        "₹30-50",
        "₹50-100",
        "> ₹100"
    ]
)

freight_review_analysis = (
    review_freight
    .dropna(subset=["freight_bucket"])
    .groupby("freight_bucket", observed=False)
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        five_star_reviews=("review_score", lambda x: (x == 5).sum()),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum()),
        avg_freight=("freight", "mean"),
        avg_freight_pct=("freight_pct", "mean")
    )
    .reset_index()
)

freight_review_analysis["five_star_rate_pct"] = (
    freight_review_analysis["five_star_reviews"]
    / freight_review_analysis["reviews"] * 100
)

freight_review_analysis["negative_review_rate_pct"] = (
    freight_review_analysis["negative_reviews"]
    / freight_review_analysis["reviews"] * 100
)

print("\nREVIEW SCORE VS FREIGHT COST")
print("-" * 70)
print(
    freight_review_analysis.to_string(index=False)
)

# --------------------------------------------------------------
# Review score by freight percentage
# --------------------------------------------------------------

review_freight["freight_pct_bucket"] = pd.cut(
    review_freight["freight_pct"],
    bins=[-np.inf, 10, 20, 30, 50, 100, np.inf],
    labels=[
        "≤ 10%",
        "10-20%",
        "20-30%",
        "30-50%",
        "50-100%",
        "> 100%"
    ]
)

freight_pct_analysis = (
    review_freight
    .dropna(subset=["freight_pct_bucket"])
    .groupby("freight_pct_bucket", observed=False)
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum())
    )
    .reset_index()
)

freight_pct_analysis["negative_review_rate_pct"] = (
    freight_pct_analysis["negative_reviews"]
    / freight_pct_analysis["reviews"] * 100
)

print("\nREVIEW SCORE VS FREIGHT % OF PRODUCT VALUE")
print("-" * 70)
print(
    freight_pct_analysis.to_string(index=False)
)
```

```python
# ==============================================================
# SECTION 14.7 — STEP 7: LOW-RATED ORDER ANALYSIS
# ==============================================================

# Merge reviews with order information
low_review = order_reviews[
    ["order_id", "review_score", "review_comment_title", "review_comment_message"]
].merge(
    orders[
        [
            "order_id",
            "customer_id",
            "order_status",
            "order_purchase_timestamp",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ],
    on="order_id",
    how="left"
)

# --------------------------------------------------------------
# Identify low-rated reviews
# --------------------------------------------------------------

low_review["rating_category"] = np.select(
    [
        low_review["review_score"] == 1,
        low_review["review_score"] == 2,
        low_review["review_score"] == 3,
        low_review["review_score"].isin([4, 5])
    ],
    [
        "1-Star",
        "2-Star",
        "3-Star",
        "4-5 Star"
    ],
    default="Unknown"
)

# --------------------------------------------------------------
# Overall low-rated review summary
# --------------------------------------------------------------

low_rating_summary = pd.DataFrame({
    "metric": [
        "Total Reviews",
        "1-Star Reviews",
        "2-Star Reviews",
        "1-2 Star Reviews",
        "1-Star Share %",
        "2-Star Share %",
        "1-2 Star Share %"
    ],
    "value": [
        len(low_review),
        (low_review["review_score"] == 1).sum(),
        (low_review["review_score"] == 2).sum(),
        low_review["review_score"].isin([1, 2]).sum(),
        (low_review["review_score"] == 1).mean() * 100,
        (low_review["review_score"] == 2).mean() * 100,
        low_review["review_score"].isin([1, 2]).mean() * 100
    ]
})

print("\nLOW-RATED REVIEW SUMMARY")
print("-" * 70)
print(low_rating_summary.to_string(index=False))

# --------------------------------------------------------------
# Review score vs order status
# --------------------------------------------------------------

order_status_reviews = (
    low_review
    .groupby("order_status")
    .agg(
        reviews=("review_score", "count"),
        avg_review_score=("review_score", "mean"),
        negative_reviews=("review_score", lambda x: x.isin([1, 2]).sum())
    )
    .reset_index()
)

order_status_reviews["negative_review_rate_pct"] = (
    order_status_reviews["negative_reviews"]
    / order_status_reviews["reviews"] * 100
)

order_status_reviews = order_status_reviews.sort_values(
    "negative_review_rate_pct",
    ascending=False
)

print("\nREVIEW PERFORMANCE BY ORDER STATUS")
print("-" * 70)
print(order_status_reviews.to_string(index=False))

# --------------------------------------------------------------
# Comment availability for low-rated reviews
# --------------------------------------------------------------

low_rated = low_review[
    low_review["review_score"].isin([1, 2])
].copy()

low_rated["has_comment"] = (
    low_rated["review_comment_message"]
    .fillna("")
    .str.strip()
    .ne("")
)

comment_summary = pd.DataFrame({
    "metric": [
        "Low-Rated Reviews",
        "Low-Rated Reviews With Comments",
        "Low-Rated Reviews Without Comments",
        "Comment Availability %"
    ],
    "value": [
        len(low_rated),
        low_rated["has_comment"].sum(),
        (~low_rated["has_comment"]).sum(),
        low_rated["has_comment"].mean() * 100
    ]
})

print("\nLOW-RATED REVIEW COMMENT ANALYSIS")
print("-" * 70)
print(comment_summary.to_string(index=False))

# --------------------------------------------------------------
# Top low-rated reviews with comments
# --------------------------------------------------------------

low_rated_comments = (
    low_rated[
        low_rated["has_comment"]
    ][
        [
            "order_id",
            "review_score",
            "review_comment_title",
            "review_comment_message"
        ]
    ]
    .sort_values("review_score")
    .head(20)
)

print("\nSAMPLE LOW-RATED REVIEWS WITH COMMENTS")
print("-" * 70)
print(low_rated_comments.to_string(index=False))
```

```python
# ==============================================================
# SECTION 14.8 — PRODUCT & CATEGORY PERFORMANCE ANALYSIS
# STEP 1: CATEGORY PERFORMANCE OVERVIEW
# ==============================================================

print("=" * 70)
print("SECTION 14.8 — PRODUCT & CATEGORY PERFORMANCE ANALYSIS")
print("=" * 70)

# --------------------------------------------------------------
# Prepare product/order data
# --------------------------------------------------------------

category_data = order_items[
    ["order_id", "product_id", "price", "freight_value"]
].merge(
    products[
        ["product_id", "product_category_name"]
    ],
    on="product_id",
    how="left"
)

# Handle missing categories
category_data["product_category_name"] = (
    category_data["product_category_name"]
    .fillna("unknown")
)

# --------------------------------------------------------------
# Category performance
# --------------------------------------------------------------

category_performance = (
    category_data
    .groupby("product_category_name")
    .agg(
        orders=("order_id", "nunique"),
        units_sold=("product_id", "count"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum"),
        unique_products=("product_id", "nunique")
    )
    .reset_index()
)

# --------------------------------------------------------------
# Derived KPIs
# --------------------------------------------------------------

category_performance["avg_order_value"] = (
    category_performance["revenue"]
    / category_performance["orders"]
)

category_performance["avg_unit_price"] = (
    category_performance["revenue"]
    / category_performance["units_sold"]
)

category_performance["avg_freight"] = (
    category_performance["freight"]
    / category_performance["orders"]
)

category_performance["freight_pct"] = np.where(
    category_performance["revenue"] > 0,
    category_performance["freight"]
    / category_performance["revenue"] * 100,
    0
)

category_performance["order_share_pct"] = (
    category_performance["orders"]
    / category_performance["orders"].sum() * 100
)

category_performance["revenue_share_pct"] = (
    category_performance["revenue"]
    / category_performance["revenue"].sum() * 100
)

category_performance = category_performance.sort_values(
    "revenue",
    ascending=False
)

# --------------------------------------------------------------
# Display
# --------------------------------------------------------------

print("\nCATEGORY PERFORMANCE OVERVIEW")
print("-" * 70)

print(
    category_performance.to_string(
        index=False,
        formatters={
            "revenue": "{:,.2f}".format,
            "freight": "{:,.2f}".format,
            "avg_order_value": "{:,.2f}".format,
            "avg_unit_price": "{:,.2f}".format,
            "avg_freight": "{:,.2f}".format,
            "freight_pct": "{:.2f}".format,
            "order_share_pct": "{:.2f}".format,
            "revenue_share_pct": "{:.2f}".format
        }
    )
)

# --------------------------------------------------------------
# Top categories by revenue
# --------------------------------------------------------------

print("\nTOP 10 CATEGORIES BY REVENUE")
print("-" * 70)

print(
    category_performance[
        [
            "product_category_name",
            "revenue",
            "units_sold",
            "orders",
            "revenue_share_pct"
        ]
    ]
    .head(10)
    .to_string(index=False)
)

# --------------------------------------------------------------
# Top categories by units sold
# --------------------------------------------------------------

print("\nTOP 10 CATEGORIES BY UNITS SOLD")
print("-" * 70)

print(
    category_performance
    .sort_values("units_sold", ascending=False)
    [
        [
            "product_category_name",
            "units_sold",
            "orders",
            "revenue",
            "avg_unit_price"
        ]
    ]
    .head(10)
    .to_string(index=False)
)

# --------------------------------------------------------------
# Category summary
# --------------------------------------------------------------

print("\nCATEGORY SUMMARY")
print("-" * 70)

print(f"Total Categories       : {category_performance['product_category_name'].nunique()}")
print(f"Total Units Sold       : {category_performance['units_sold'].sum():,.0f}")
print(f"Total Revenue          : {category_performance['revenue'].sum():,.2f}")
print(f"Total Freight          : {category_performance['freight'].sum():,.2f}")

print(
    f"Average Category Revenue : "
    f"{category_performance['revenue'].mean():,.2f}"
)
```

```python
# ======================================================================
# SECTION 14.9 — INDIVIDUAL PRODUCT PERFORMANCE ANALYSIS
# ======================================================================

print("=" * 70)
print("SECTION 14.9 — INDIVIDUAL PRODUCT PERFORMANCE ANALYSIS")
print("=" * 70)

# Product performance overview
product_overview = pd.DataFrame({
    "metric": [
        "Total Unique Products Sold",
        "Total Units Sold",
        "Total Revenue",
        "Total Freight",
        "Average Revenue per Product",
        "Average Units per Product",
        "Average Freight per Product"
    ],
    "value": [
        order_items["product_id"].nunique(),
        order_items["order_item_id"].count(),
        order_items["price"].sum(),
        order_items["freight_value"].sum(),
        order_items.groupby("product_id")["price"].sum().mean(),
        order_items.groupby("product_id")["order_id"].count().mean(),
        order_items.groupby("product_id")["freight_value"].sum().mean()
    ]
})

print("\nPRODUCT PERFORMANCE OVERVIEW")
print("-" * 70)
print(product_overview.to_string(index=False))
```

```python
# ================================================================
# SECTION 14.9 — STEP 2: PREPARE PRODUCT PERFORMANCE
# ================================================================

# Check available columns
print("order_items columns:")
print(order_items.columns.tolist())

print("\nproducts columns:")
print(products.columns.tolist())
```

```python
# ================================================================
# SECTION 14.9 — STEP 2: PREPARE PRODUCT PERFORMANCE
# ================================================================

# Merge product category information into order_items
order_items_product = order_items.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

# Handle missing categories
order_items_product["product_category_name"] = (
    order_items_product["product_category_name"]
    .fillna("unknown")
)

# Create product-level performance table
product_performance = (
    order_items_product
    .groupby(
        ["product_id", "product_category_name"],
        as_index=False
    )
    .agg(
        orders=("order_id", "nunique"),
        units_sold=("order_item_id", "count"),
        revenue=("price", "sum"),
        freight=("freight_value", "sum")
    )
)

# Average selling price per unit
product_performance["avg_unit_price"] = (
    product_performance["revenue"] /
    product_performance["units_sold"]
)

# Freight as percentage of product revenue
product_performance["freight_pct"] = (
    product_performance["freight"] /
    product_performance["revenue"] * 100
)

# Sort by revenue
product_performance = product_performance.sort_values(
    ["revenue", "units_sold"],
    ascending=[False, False]
).reset_index(drop=True)

print("\n" + "=" * 70)
print("PRODUCT PERFORMANCE TABLE CREATED")
print("=" * 70)

print(
    product_performance.head(10).to_string(index=False)
)

print("\nShape:", product_performance.shape)
print(
    "Unique Products:",
    product_performance["product_id"].nunique()
)
```

```python
# ================================================================
# SECTION 14.9 — STEP 3: TOP 20 PRODUCTS BY UNITS SOLD
# ================================================================

top_products_units = (
    product_performance
    .sort_values(
        ["units_sold", "revenue"],
        ascending=[False, False]
    )
    .head(20)
)

print("\n" + "=" * 70)
print("TOP 20 PRODUCTS BY UNITS SOLD")
print("=" * 70)

print(
    top_products_units[
        [
            "product_id",
            "product_category_name",
            "units_sold",
            "orders",
            "revenue",
            "avg_unit_price"
        ]
    ].to_string(index=False)
)
```

```python
# ================================================================
# SECTION 14.9 — STEP 4: TOP 20 PRODUCTS BY REVENUE
# ================================================================

top_products_revenue = (
    product_performance
    .sort_values(
        ["revenue", "units_sold"],
        ascending=[False, False]
    )
    .head(20)
)

print("\n" + "=" * 70)
print("TOP 20 PRODUCTS BY REVENUE")
print("=" * 70)

print(
    top_products_revenue[
        [
            "product_id",
            "product_category_name",
            "revenue",
            "units_sold",
            "orders",
            "avg_unit_price",
            "freight",
            "freight_pct"
        ]
    ].to_string(index=False)
)
```

```python
# ================================================================
# SECTION 14.9 — STEP 4: PRODUCT PERFORMANCE SEGMENTS
# ================================================================

# Calculate overall medians
median_units = product_performance["units_sold"].median()
median_revenue = product_performance["revenue"].median()

# Segment products based on units sold and revenue
product_performance["performance_segment"] = "Low Volume / Low Revenue"

product_performance.loc[
    (product_performance["units_sold"] >= median_units) &
    (product_performance["revenue"] < median_revenue),
    "performance_segment"
] = "High Volume / Low Revenue"

product_performance.loc[
    (product_performance["units_sold"] < median_units) &
    (product_performance["revenue"] >= median_revenue),
    "performance_segment"
] = "Low Volume / High Revenue"

product_performance.loc[
    (product_performance["units_sold"] >= median_units) &
    (product_performance["revenue"] >= median_revenue),
    "performance_segment"
] = "High Volume / High Revenue"


print("\n" + "=" * 70)
print("PRODUCT PERFORMANCE SEGMENTS")
print("=" * 70)

segment_summary = (
    product_performance
    .groupby("performance_segment", as_index=False)
    .agg(
        products=("product_id", "count"),
        units_sold=("units_sold", "sum"),
        revenue=("revenue", "sum"),
        avg_revenue=("revenue", "mean"),
        avg_units_sold=("units_sold", "mean")
    )
    .sort_values("revenue", ascending=False)
)

print(segment_summary.to_string(index=False))
```

```python
# ================================================================
# SECTION 14.9 — STEP 6: PRODUCT REVENUE CONCENTRATION
# ================================================================

product_revenue_rank = (
    product_performance
    .sort_values("revenue", ascending=False)
    .reset_index(drop=True)
)

product_revenue_rank["cumulative_revenue"] = (
    product_revenue_rank["revenue"].cumsum()
)

total_product_revenue = product_revenue_rank["revenue"].sum()

product_revenue_rank["cumulative_revenue_pct"] = (
    product_revenue_rank["cumulative_revenue"]
    / total_product_revenue * 100
)

print("\n" + "=" * 70)
print("PRODUCT REVENUE CONCENTRATION")
print("=" * 70)

# Revenue generated by top 10, 50, 100, 500 and 1000 products
concentration_rows = []

for n in [10, 50, 100, 500, 1000]:
    revenue = product_revenue_rank.head(n)["revenue"].sum()
    
    concentration_rows.append({
        "top_products": n,
        "revenue": revenue,
        "revenue_share_pct": revenue / total_product_revenue * 100
    })

concentration_summary = pd.DataFrame(concentration_rows)

print(concentration_summary.to_string(index=False))
```

```python
# ================================================================
# PRODUCTS REQUIRED TO GENERATE 50%, 80% AND 90% OF REVENUE
# ================================================================

thresholds = [50, 80, 90, 95]

print("\n" + "=" * 70)
print("REVENUE CONCENTRATION THRESHOLDS")
print("=" * 70)

for threshold in thresholds:
    
    products_needed = (
        product_revenue_rank[
            product_revenue_rank["cumulative_revenue_pct"] >= threshold
        ]
        .index[0] + 1
    )
    
    print(
        f"{threshold}% of revenue requires "
        f"{products_needed:,} products"
    )
```

```python
# ================================================================
# SECTION 14.10 — STEP 1A: VERIFY PAYMENT TABLE
# ================================================================

print("=" * 70)
print("FACT PAYMENTS COLUMNS")
print("=" * 70)

print(fact_payments.columns.tolist())

print("\nShape:", fact_payments.shape)

print("\nSample:")
print(fact_payments.head().to_string(index=False))
```

```python
# ================================================================
# SECTION 14.10 — STEP 1B: PAYMENT TYPE VS ORDER VALUE
# ================================================================

payment_order_analysis = (
    fact_payments
    .groupby("payment_type", as_index=False)
    .agg(
        orders=("order_id", "nunique"),
        payments=("payment_value", "count"),
        total_payment=("payment_value", "sum"),
        avg_payment_value=("payment_value", "mean"),
        median_payment_value=("payment_value", "median"),
        avg_installments=("payment_installments", "mean")
    )
)

# Share of unique orders
payment_order_analysis["order_share_pct"] = (
    payment_order_analysis["orders"]
    / payment_order_analysis["orders"].sum()
    * 100
)

# Share of payment value
payment_order_analysis["payment_value_share_pct"] = (
    payment_order_analysis["total_payment"]
    / payment_order_analysis["total_payment"].sum()
    * 100
)

payment_order_analysis = payment_order_analysis.sort_values(
    "total_payment",
    ascending=False
)

print("\n" + "=" * 70)
print("SECTION 14.10 — PAYMENT TYPE VS ORDER VALUE")
print("=" * 70)

print(
    payment_order_analysis.to_string(index=False)
)
```

```python
# ================================================================
# SECTION 14.10 — STEP 2: INSTALLMENTS VS PAYMENT VALUE
# ================================================================

installment_analysis = (
    fact_payments
    .groupby("payment_installments", as_index=False)
    .agg(
        payments=("payment_value", "count"),
        total_payment=("payment_value", "sum"),
        avg_payment_value=("payment_value", "mean"),
        median_payment_value=("payment_value", "median")
    )
)

installment_analysis["payment_value_share_pct"] = (
    installment_analysis["total_payment"]
    / installment_analysis["total_payment"].sum()
    * 100
)

installment_analysis = installment_analysis.sort_values(
    "payment_installments"
)

print("\n" + "=" * 70)
print("SECTION 14.10 — INSTALLMENT VS PAYMENT VALUE")
print("=" * 70)

print(
    installment_analysis.to_string(index=False)
)
```

```python
# ================================================================
# CHECK KEY DATAFRAMES
# ================================================================

tables_to_check = [
    "powerbi_fact_payments",
    "fact_payments",
    "product_performance",
    "order_items",
    "products",
    "customers",
    "sellers"
]

for table_name in tables_to_check:

    if table_name in globals():

        df = globals()[table_name]

        print("\n" + "-" * 70)
        print(f"{table_name}")
        print("-" * 70)
        print("Shape:", df.shape)
        print("Columns:")
        print(df.columns.tolist())
```

```python
# ================================================================
# SECTION 16 — STEP 2: VALIDATE POWER BI MODEL KEYS
# ================================================================

print("\n" + "=" * 70)
print("POWER BI MODEL KEY VALIDATION")
print("=" * 70)


# ------------------------------------------------
# DIMENSION PRIMARY KEY CHECKS
# ------------------------------------------------

dimension_keys = {
    "dim_customer": "customer_id",
    "dim_product": "product_id",
    "dim_seller": "seller_id",
    "dim_order": "order_id",
    "dim_date": "date"
}

for table, key in dimension_keys.items():

    if table in globals():

        df = globals()[table]

        if key in df.columns:

            total = len(df)
            unique = df[key].nunique()
            nulls = df[key].isna().sum()
            duplicates = total - unique

            print(f"\n{table}")
            print(f"Key: {key}")
            print(f"Rows: {total:,}")
            print(f"Unique keys: {unique:,}")
            print(f"Null keys: {nulls:,}")
            print(f"Duplicate keys: {duplicates:,}")


# ------------------------------------------------
# FACT TABLE KEY CHECKS
# ------------------------------------------------

print("\n" + "=" * 70)
print("FACT TABLE KEY CHECKS")
print("=" * 70)

fact_keys = {
    "fact_order_items": [
        "order_id",
        "product_id",
        "seller_id"
    ],

    "fact_payments": [
        "order_id"
    ],

    "fact_reviews": [
        "order_id"
    ]
}

for table, keys in fact_keys.items():

    if table in globals():

        df = globals()[table]

        print(f"\n{table}")
        print(f"Rows: {len(df):,}")

        for key in keys:

            if key in df.columns:

                print(
                    f"  {key}: "
                    f"unique={df[key].nunique():,}, "
                    f"nulls={df[key].isna().sum():,}"
                )
```

```python
# ================================================================
# SECTION 16 — STEP 3: FINAL POWER BI TABLE CHECK
# ================================================================

powerbi_tables = [
    "dim_customer",
    "dim_product",
    "dim_seller",
    "dim_order",
    "dim_date",
    "fact_order_items",
    "fact_payments",
    "fact_reviews"
]

print("\n" + "=" * 70)
print("FINAL POWER BI TABLES")
print("=" * 70)

for table_name in powerbi_tables:

    df = globals()[table_name]

    print(f"\n{table_name}")
    print("-" * 50)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())


# ================================================================
# CHECK CUSTOMER KEY
# ================================================================

print("\n" + "=" * 70)
print("CUSTOMER KEY CHECK")
print("=" * 70)

print(dim_customer.head())
print("\nUnique customer_id:",
      dim_customer["customer_id"].nunique()
      if "customer_id" in dim_customer.columns
      else "customer_id NOT FOUND")

print("Unique customer_unique_id:",
      dim_customer["customer_unique_id"].nunique()
      if "customer_unique_id" in dim_customer.columns
      else "customer_unique_id NOT FOUND")


# ================================================================
# CHECK ORDER DATE COLUMNS
# ================================================================

print("\n" + "=" * 70)
print("ORDER DATE COLUMNS")
print("=" * 70)

date_columns = [
    col for col in dim_order.columns
    if "date" in col.lower()
]

print(date_columns)


# ================================================================
# CHECK DATE RANGE
# ================================================================

for col in date_columns:

    print(f"\n{col}")

    try:
        dates = pd.to_datetime(dim_order[col], errors="coerce")

        print("Min:", dates.min())
        print("Max:", dates.max())
        print("Nulls:", dates.isna().sum())

    except Exception as e:
        print("Could not process:", e)
```
